## 3-카 — 모델 예측이 실제와 몇 % 달랐나: 평소 해(2024 · 2025년)와 2026년 1 ~ 6월

**표의 숫자**

- 모델이 예측한 것은 전력 수요가 아니라 시간별 LNG 발전량입니다. 그래서 이 셀의 %는 LNG 발전량 예측이 실제와 몇 % 달랐나입니다.
- [표 1]은 해마다 그 전 해까지를 배운 모델이 처음 보는 해를 예측했을 때입니다. 2024 · 2025년은 평소 해이고, 2026년 1 ~ 6월은 비교한 해라 평소와 달랐던 만큼도 들어 있습니다.
- 「한 시각에 틀린 크기 평균 (%)」는 시각마다 |실제 − 모델 예측| ÷ 실제 × 100을 구해 그 해 모든 시각으로 평균 낸 것입니다(+ · −를 떼고 크기만). 평균 절대 백분율 오차(MAPE)라고 부르고, 100에서 빼서 「정확도 몇 %」로 말하기도 합니다.
- 「한 달 평균으로 틀린 크기 평균 (%)」는 달마다 (실제 − 모델 예측) ÷ 실제 × 100을 그 달 한 시각 평균으로 구하고, + · −를 떼고 달끼리 평균 낸 것입니다. 9단계에서 평소 범위를 정한 단위(한 달 평균)와 같습니다. 한 달 안에서 + 인 시각과 − 인 시각이 상쇄되어 보통 한 시각 단위 %보다 작게 나옵니다.
- [표 2] 「실제 − 모델 예측」은 + 면 실제가 모델 예측보다 많았다는 뜻입니다. 「평소 범위 (± %)」는 평소 범위(2024 · 2025년에 모델이 한 달 평균으로 가장 크게 틀린 크기, MWh)를 그 달 실제로 나눈 것이라, 「실제 − 모델 예측 (%)」의 크기가 이보다 작으면 평소 범위 안입니다.
- 표 아래 「계산 예」 두 줄은 실제 값으로 한 번씩 계산해 보이고, 마지막 줄은 이 셀의 모델이 7 · 9단계와 같은지 봅니다(2025년 결정계수와 한 시각에 틀린 크기(MWh)는 7단계 2-가, [표 2]의 MWh 세 칸은 9단계 2-가와 같아야 합니다).

- 이 셀은 판정을 새로 하지 않습니다. 9단계에서 MWh로 한 판정(달마다 「실제 − 모델 예측」의 크기가 평소 범위 안인가)을 %로 옮겨 읽는 셀이라, [표 2]에서 평소 범위 안으로 읽히는 달이 9단계 2-가 「그 크기를 평소 범위와 비교」 칸과 다르면 계산이 틀린 것입니다.
- 「어느 정도에서 어느 정도 벗어났나」는 [표 1]의 평소 해 두 줄(모델이 평소에 몇 % 틀리나)과 [표 2]의 2026년 달마다 %를 나란히 놓고 읽습니다.
- README에 적을 때는 한 시각 단위 %와 한 달 단위 %를 섞지 않고, 어느 단위인지 붙여 적습니다.

```python
# 11단계 결론 · 3-카
import os, glob                                                                 # 폴더와 파일 경로를 다루는 기본 모듈 · 파일 이름을 한꺼번에 찾는 모듈
import numpy as np                                                              # 숫자 계산
import pandas as pd                                                             # 표(데이터프레임)를 다루는 라이브러리
from IPython.display import display                                             # 표를 글자가 아니라 격자로 그린다
from sklearn.ensemble import HistGradientBoostingRegressor                      # 8단계에서 고른 그래디언트 부스팅

ROOT = os.getcwd()                                                              # 지금 폴더(보통 notebooks/)에서 시작해서
while not os.path.isdir(os.path.join(ROOT, "data")) and ROOT != os.path.dirname(ROOT):
    ROOT = os.path.dirname(ROOT)                                                # data 폴더가 보일 때까지 위로 = 프로젝트 루트
RAW = os.path.join(ROOT, "data", "raw")                                         # 2단계에서 받아 둔 원자료
HOT = os.path.join(ROOT, "data", "hot")                                         # 5단계가 만들어 둔 표

# ── 1. 9단계와 같은 표 · 같은 입력 7열 · 같은 설정 ──
tbl = pd.read_csv(os.path.join(HOT, "features_2023_2024_2025.csv"), encoding="utf-8-sig")   # 5단계 「저장」 셀이 만든 표
tbl["ts"] = pd.to_datetime(tbl["ts"])                                          # 글자로 저장된 시각을 시각 자료형으로
tbl = tbl.set_index("ts").sort_index()                                          # 시각을 줄 이름으로 두고 시간 순서대로
전체 = tbl.dropna(subset=["3시간 전 기온"])                                       # 3시간 전 기온이 빈 9시각은 뺀다 — 6 ~ 9단계와 같다
y = 전체["lng_mwh"]                                                             # 예측 대상 — 그 시각의 LNG 발전량(MWh)
X = 전체[["hour24", "month", "icsr", "ta", "dow", "is_holiday", "명절"]]          # 입력 7열 — 몇 시 · 몇 월 · 일사 · 기온 · 요일 · 공휴일 · 명절
def 그래디언트부스팅():                                                            # 8단계에서 확정한 설정 그대로
    return HistGradientBoostingRegressor(early_stopping=False, random_state=0)

# ── 2. 평소 해 — 그 전 해까지를 배워 2024 · 2025년을 예측한다(9단계 준비 ②와 같다) ──
비교 = {}                                                                        # 해마다 시각마다 실제와 모델 예측
for 해, 배운해 in [(2024, [2023]), (2025, [2023, 2024])]:
    배울곳 = 전체["year"].isin(배운해); 맞힐곳 = 전체["year"] == 해                     # 배우는 시각 · 예측하는 시각
    모델 = 그래디언트부스팅().fit(X[배울곳], y[배울곳])                                # 그 전 해까지를 배운다
    비교[해] = pd.DataFrame({"실제": y[맞힐곳], "모델 예측": 모델.predict(X[맞힐곳])})

# ── 3. 2026년 1 ~ 6월 — 입력 7열은 9단계 준비 ③, 실제 LNG는 9단계 2-가와 같은 방법 ──
기상 = pd.concat([pd.read_csv(f, encoding="utf-8-sig", usecols=["tm", "ta", "icsr"])      # 서울 ASOS — 시각 · 기온 · 일사만 읽는다
                  for f in sorted(glob.glob(os.path.join(RAW, "asos", "asos_108_2026*.csv")))], ignore_index=True)
기상.index = pd.to_datetime(기상["tm"])                                            # '2026-01-01 00:00' 글자를 시각으로
입력26 = 기상.sort_index().loc["2026-01-01 00:00":"2026-06-30 23:00", ["ta", "icsr"]].copy()   # 비교 기간 1 ~ 6월
입력26["icsr"] = 입력26["icsr"].fillna(0)                                        # 해가 없을 때 비는 일사는 0 — 3단계와 같다
입력26["month"] = 입력26.index.month                                               # 1 ~ 6월
입력26["hour24"] = 입력26.index.hour                                               # 0 ~ 23시
입력26["dow"] = 입력26.index.dayofweek                                             # 요일 0=월 … 6=일
휴일 = pd.read_csv(os.path.join(RAW, "holidays_kr.csv"), encoding="utf-8-sig")    # 2단계 `02_수집_공휴일`이 받아 둔 파일
날짜 = pd.Series(입력26.index.date, index=입력26.index)                            # 시각마다 그 날짜
입력26["is_holiday"] = 날짜.isin(set(pd.to_datetime(휴일["date"]).dt.date)).astype(int)   # 공휴일이면 1 — 5단계와 같다
명절날 = set(pd.to_datetime(휴일.loc[휴일["name"].str.contains("설날|추석"), "date"]).dt.date)   # 이름에 설날 · 추석이 든 날
입력26["명절"] = 날짜.isin(명절날).astype(int)                                      # 설 · 추석 연휴면 1 — 5단계와 같다

실제 = []                                                                       # 날마다 24시각의 실제 LNG
for f in sorted(glob.glob(os.path.join(RAW, "api", "2026*.csv"))):             # 2026년 하루치 파일을 날짜 순서대로
    하루 = pd.to_datetime(os.path.basename(f)[:8])                               # 파일 이름의 날짜(20260101)
    if 하루.month > 6 or os.path.getsize(f) < 100: continue                       # 1 ~ 6월만 · 0행으로 온 날은 건너뛴다
    t = pd.read_csv(f, encoding="utf-8-sig")                                     # 그날 파일을 읽는다
    if len(t) != 192: continue                                                    # 연료 여덟 × 24시간이 다 온 날만 — 5월 2 ~ 17일이 빠진다
    t = t[t["fuelTpCd"] == "LNG"]                                                # LNG 줄만
    실제.append(pd.Series(t["amgo"].values, index=하루 + pd.to_timedelta(t["tradeNo"].astype(int) - 1, unit="h")))   # 1 ~ 24시를 0 ~ 23시 시각으로
실제 = pd.concat(실제).sort_index()                                               # 시각마다 실제 LNG 발전량(MWh)
모델 = 그래디언트부스팅().fit(X, y)                                                  # 2023 ~ 2025년 세 해를 배운 모델 — 6단계 물음 5에서 정한 것
비교[2026] = pd.DataFrame({"실제": 실제.values,
                           "모델 예측": 모델.predict(입력26.reindex(실제.index)[X.columns])}, index=실제.index)

# ── 4. 몇 % 틀렸나 ──
def 달마다(d):                                                                   # 달마다 한 시각 평균
    g = d.groupby(d.index.month)[["실제", "모델 예측"]].mean()
    g["실제 − 모델 예측"] = g["실제"] - g["모델 예측"]                             # + 면 실제가 모델 예측보다 많았다
    g["%"] = g["실제 − 모델 예측"] / g["실제"] * 100                                # 그 크기가 그 달 실제의 몇 %인가
    return g
월 = {해: 달마다(d) for 해, d in 비교.items()}
평소범위 = max(월[2024]["실제 − 모델 예측"].abs().max(), 월[2025]["실제 − 모델 예측"].abs().max())   # 2024 · 2025년에 한 달 평균으로 가장 크게 틀린 크기(9단계 1-다)

이름 = {2024: "2024년 — 2023년으로 배워 예측(평소 해)", 2025: "2025년 — 2023 · 2024년으로 배워 예측(평소 해 · 7단계 채점)",
        2026: "2026년 1 ~ 6월 — 2023 ~ 2025년으로 배워 예측(비교한 해)"}
표1 = []
for 해, d in 비교.items():
    차 = d["실제"] - d["모델 예측"]                                                  # 시각마다 실제 − 모델 예측
    m = 월[해]; k = m["%"].abs().idxmax()                                           # 한 달 평균으로 가장 크게 틀린 달
    표1.append({"예측한 해": 이름[해], "시각 수": f"{len(d):,}",
                "결정계수": f"{1 - (차 ** 2).sum() / ((d['실제'] - d['실제'].mean()) ** 2).sum():.4f}",   # 6 ~ 9단계와 같은 식
                "한 시각에 틀린 크기 평균 (MWh)": f"{차.abs().mean():,.0f}",                               # 평균 절대 오차
                "한 시각에 틀린 크기 평균 (%)": f"{(차.abs() / d['실제'] * 100).mean():.1f}",              # 시각마다 |실제 − 모델 예측| ÷ 실제 × 100의 평균
                "한 달 평균으로 틀린 크기 평균 (%)": f"{m['%'].abs().mean():.1f}",                        # 달마다 %의 크기를 달끼리 평균
                "한 달 평균으로 가장 크게 틀린 달 (% · + 면 실제가 많았다)": f"{k}월 {m.loc[k, '%']:+.1f}"})
표1 = pd.DataFrame(표1).set_index("예측한 해")

m = 월[2026]
날수 = pd.Series(비교[2026].index.normalize(), index=비교[2026].index).groupby(비교[2026].index.month).nunique()   # 달마다 비교한 날 수
표2 = pd.DataFrame({"비교한 날": 날수,
                    "실제 (MWh)": m["실제"].map("{:,.0f}".format),
                    "모델 예측 (MWh)": m["모델 예측"].map("{:,.0f}".format),
                    "실제 − 모델 예측 (MWh)": m["실제 − 모델 예측"].map("{:+,.0f}".format),
                    "실제 − 모델 예측 (%)": m["%"].map("{:+.1f}".format),
                    "평소 범위 (± %)": (평소범위 / m["실제"] * 100).map("{:.1f}".format)})   # 평소 범위 MWh ÷ 그 달 실제 × 100
표2.index = [f"{k}월" for k in 표2.index]                                         # 줄 이름 1 → 1월

print("[표 1] 모델 예측이 실제와 얼마나 달랐나 — 그래디언트 부스팅(입력 7열), 해마다 그 전 해까지를 배워 예측")
print("       %는 + · −를 떼고 크기만 본 것 · 「가장 크게 틀린 달」만 + · −를 붙였다")
display(표1)
print(f"[표 2] 2026년 달마다 한 시각 평균 — 실제 − 모델 예측은 + 면 실제가 많았다 · 평소 범위(± %)는 {평소범위:,.0f} MWh ÷ 그 달 실제 × 100")
display(표2)

h = 비교[2025].index[0]; a, b = 비교[2025].loc[h, "실제"], 비교[2025].loc[h, "모델 예측"]   # 2025년 첫 시각으로 한 번 계산해 보인다
print(f"계산 예 · 한 시각 — {h.year}년 {h.month}월 {h.day}일 {h.hour}시: |실제 {a:,.0f} − 모델 예측 {b:,.0f}| ÷ {a:,.0f} × 100 = {abs(a - b) / a * 100:.1f}% "
      f"→ 그 해 모든 시각의 평균이 [표 1] 「한 시각에 틀린 크기 평균 (%)」")
j = m.loc[1]                                                                       # 2026년 1월로 한 번 계산해 보인다
print(f"계산 예 · 한 달 — 2026년 1월: (실제 {j['실제']:,.0f} − 모델 예측 {j['모델 예측']:,.0f}) ÷ {j['실제']:,.0f} × 100 = {j['%']:+.1f}% "
      f"· 평소 범위 {평소범위:,.0f} ÷ {j['실제']:,.0f} × 100 = {평소범위 / j['실제'] * 100:.1f}%")
같다 = (표1.iloc[0]["결정계수"] == "0.7807" and 표1.iloc[1]["결정계수"] == "0.7553" and 표1.iloc[1]["한 시각에 틀린 크기 평균 (MWh)"] == "2,364"
        and f"{평소범위:,.0f}" == "2,177" and list(표2["모델 예측 (MWh)"]) == ["21,613", "19,199", "18,988", "16,918", "16,068", "16,582"])
print("모델이 7 · 9단계와 같은지 — 결정계수(2024년 0.7807 · 2025년 0.7553) · 2025년 한 시각에 틀린 크기(2,364 MWh) · 평소 범위(2,177 MWh) · "
      f"2026년 달마다 모델 예측(9단계 2-가)이 {'같다' if 같다 else '다르다(scikit-learn 판을 확인해 주세요)'}")
```
