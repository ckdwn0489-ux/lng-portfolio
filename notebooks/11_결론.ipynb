| 물음 | 결과 | 결정 |
|---|---|---|
| 1 — 이 프로젝트의 답은 무엇인가 | 2026년 1 ~ 6월 LNG 발전량은 달마다 평소 범위 안이라 달랐다고 말할 근거가 없다. 원전 발전량이 모델 예측보다 적었던 것과 석탄 발전량이 많았던 것 · 전력 수요가 적었던 것이 상쇄됐고, 석탄이 많았던 까닭은 찾지 못했다. 처음 짐작 여섯 줄은 맞지 않은 것이 더 많았다 | README 「결과」와 「처음 짐작과 데이터」에 넣을 글(10-30, 1-가 · 1-나 아래 「✍️ 정할 것」) |
| 2 — README에 남길 한계 | 앞 단계에서 넘어온 한계는 모두 남긴다. 「한계」에 없던 경계 문단과 열두 줄을 넣고, 이미 있던 줄은 「결과」 글과 같은 말로 고쳤다(경계 문단과 마흔여섯 줄) | README 「한계」(10-30, 2-가 — 맡기심 · 「좋아 이정도면」) |

**지금** — 물음 3(README 전체를 처음 보는 사람이 읽을 판으로 — 남에게 보여 줄 파일 · 3분 설명)을 열었습니다(10-30). 3-가(README 새 판의 차례)는 맡기셔서 클로드가 정했고, 3-나(전체 흐름 — 단계마다 한 일과 정한 것) · 3-다(문제) · 3-라(데이터) · 3-마(방법) · 3-바(결과를 읽는 법) · 3-사(그림 둘과 설명 줄) · 3-아(「한계」를 넷으로 묶은 글)를 정했습니다. 핵심본은 둘로 나눕니다 — README 맨 위 「한눈에 보기」(처음 보는 사람이 읽는 3분짜리)와 창주 님이 전체를 이해하시는 10분짜리입니다. 3-자(한눈에 보기) · 3-차(파일 안내)까지 정했습니다. 다음은 README의 진행 기록을 `docs/진행기록.md`로 옮기는 일과 공개할 판 고르기입니다.

# 11. 결론 — 전체 11단계 중 11단계

**이 단계의 목적** — 1단계에서 세운 물음(2026년 LNG 발전량은 평소 패턴에서 얼마나 벗어났나, 그 차이의 주요 원인은 무엇인가)에 9 · 10단계까지 쌓아 온 결론으로 답하고, 처음 보는 사람이 읽을 수 있게 README를 문제 → 데이터 → 방법 → 결과 → 한계로 다시 쓰는 단계입니다. 새로 분석해 새 결과를 내지 않고, 모델도 고치지 않습니다.

**이 노트북을 연 흐름** — 10-29에 10단계를 끝내고 커밋한 뒤, 창주 님이 「lng 11단계 시작해보자」라고 하셨습니다. 지침 11단계 칸의 노트북 칸은 「README」라서, 정하실 것을 적는 칸은 이 노트북에 두고 README는 여기 적으신 말로 다시 쓰기로 했습니다(클로드 판단 — 어느 파일에 둘지는 클로드가 정하는 일입니다). 물음 1의 틀(앞에서 정한 것 · 이 물음이 정할 것 · 뒤에서 할 것)을 채팅으로 드렸고, 창주 님이 그 물음이 정하는 것을 한 줄로 적으신 뒤 이 파일을 만들었습니다. 그 문답은 물음 1 머리 칸 아래에 옮겼습니다.

## 데이터와 모델

**데이터가 넘어온 길**

1. 1단계 — 물음을 정했습니다. 2026년 LNG 발전량은 평소 패턴에서 얼마나 벗어났나, 그 차이의 주요 원인은 무엇인가. 맞힐 값은 시간별 LNG 발전량(MWh)입니다.
2. 2 · 3단계 — 전력거래소와 기상청 자료에서 시간마다 LNG · 원전 발전량과 기온 · 햇빛 같은 값을 받고, 값이 잘못된 기간에 까닭을 붙여 뺐습니다 — 2024년 2월 29일(API에 그날이 없다) · 2024년 8월(API 기록이 잘못 들어 있다)은 학습에서, 2026년 5월 2 ~ 17일(값이 잘못돼 있다)은 비교에서 뺐습니다.
3. 4 · 5단계 — 무엇이 LNG를 움직이는지 보고, 모델이 쓸 열을 파일 하나(`data/hot/features_2023_2024_2025.csv`)로 저장했습니다.
4. 6단계 — 입력을 정했습니다. 원전 · 석탄 · 가격 · 총 전력거래량 · 습도는 넣지 않았습니다. 2023 · 2024년으로 배우고, 2025년으로 채점하고, 2026년 1 ~ 6월과 비교하기로 했습니다.
5. 7 · 8단계 — 2025년으로 채점해 모델 넷 모두 기준선보다 나았고, 그래디언트 부스팅을 골랐습니다.
6. 9단계 — 그래디언트 부스팅을 2023 ~ 2025년으로 다시 배워 2026년 1 ~ 6월을 맞히고, 실제와 비교했습니다.
7. 10단계 — 그 차이를 배 몇 척으로 바꾸고, 석탄이 모델 예측보다 많았던 까닭을 찾아보고(LNG와 석탄의 연료비단가 차이로는 설명되지 않았고, 시기가 맞는 후보 셋 가운데 무엇이 까닭인지는 가를 수 없었다), 이 프로젝트가 본 것과 보지 못한 것의 경계를 그림 둘로 그렸습니다.
8. 11단계(이 노트북) — 9 · 10단계의 결론으로 1단계 물음에 답하고, README를 다시 씁니다.

**모델의 입력** (트리 계열 7열 — 6단계에서 정한 그대로)

- 몇 시인가 · 몇 월인가
- 무슨 날인가 — 요일 · 공휴일인가 · 명절인가

- 넣지 않는 것 — 원전 · 석탄 · 총 전력거래량 · 가격 · 습도

**2026년 LNG를 쓰는 곳** — 이 노트북에는 코드가 없어서 2026년 LNG를 새로 읽지 않습니다. 결론에 쓰는 숫자는 모두 앞 단계 📌 결론 칸 · 닫는 칸과 거기서 가리키는 셀 출력(2단계 물음 2 · 4단계 2-마 같은 것)에 있는 것입니다.

## 📌 물음 1 결론

**왜 생긴 물음인가** — 1단계에서 세운 물음(2026년 LNG 발전량은 평소 패턴에서 얼마나 벗어났나, 그 차이의 주요 원인은 무엇인가)에 9 · 10단계 결론으로 답을 적은 적이 없었습니다. 그 답을 README 「결과」에 넣을 글로 적고, 1단계에 짐작한 원인 여섯 줄 옆에 데이터가 보여 준 것을 놓는 물음입니다.

### 결정 — 2026년 1 ~ 6월 LNG 발전량은 달마다 평소 범위 안이라 평소 패턴과 달랐다고 말할 근거가 없다. 그 안에서 원전 발전량이 모델 예측보다 적었던 것과 석탄 발전량이 많았던 것 · 전력 수요가 적었던 것이 상쇄됐고, 석탄이 많았던 까닭은 찾지 못했다. 처음 짐작 여섯 줄은 맞지 않은 것이 더 많았다

**같이 정한 것**

1. README 「결과」에 넣을 글 — 1-가 아래 「✍️ 정할 것」(10-30 「오키 좋아」). README 「결과」에 같은 글을 넣었다.
2. 짐작 ①(LNG는 같은 열을 낼 때 석탄보다 이산화탄소를 적게 내니, LNG를 쓰는 비중이 조금씩 커질 것이다)의 판정 — 세 해의 전력거래량에서 LNG가 차지한 비율은 2023년보다 커졌지만 해마다 늘지는 않았다. 2026년 1 ~ 6월은 그 비율을 내지 않고 모델 예측과 비교했다. 석탄과 LNG의 발전량이 모델 예측보다 많았던 양 가운데 1월은 거의 다 LNG라 짐작과 같은 쪽이었고, 2월부터는 석탄의 비율이 커져 짐작과 다른 쪽이었다. 이산화탄소 때문인지는 볼 자료가 없다(1-나 아래 「✍️ 정할 것」 1 · 2, 10-30 「응 이정도면 뭐 괜찮은거 같은데?」).
3. README 「처음 짐작과 데이터」에 넣을 글 — 짐작 여섯 가운데 맞은 것이 하나(② 추우면 난방 수요가 늘어 LNG를 많이 쓸 것이다), 반대였던 것이 하나(③ 발전용 LNG는 계절에 크게 영향받지 않을 것이다 — 실제로는 계절에 따라 크게 달랐다), 「어느 연료가 더 싼가」로는 2026년 1 ~ 6월에 맞지 않은 것이 둘(④ 연료별 효율과 가격, 경제성이 무엇을 쓸지 정한다 · ⑥ 상대적인 가격과 에너지원마다의 특성이 LNG 발전량을 정한다), 일부만 맞은 것이 둘(① 이산화탄소 · ⑤ 카타르에서 가져오는 LNG가 줄면 석탄이나 원자력에서 가져올 것이다)이다(1-나 아래 「✍️ 정할 것」 3, 10-30 「좋아」).

**이 물음에서 알게 된 것** — 앞 단계 결론을 한 글로 이으니, 2026년 1 ~ 6월에 대해 말할 수 있는 것과 말할 수 없는 것이 갈렸습니다. 말할 수 있는 것은 LNG 발전량이 달마다 평소 범위 안이었다는 것, 원전 발전량이 모델 예측보다 적었던 것과 석탄 발전량이 많았던 것 · 전력 수요가 적었던 것이 상쇄됐다는 것, 석탄 발전량이 많았던 것은 LNG와 석탄의 연료비단가 차이로 설명되지 않는다는 것입니다. 말할 수 없는 것은 석탄 발전량이 많았던 까닭, LNG가 모자랐는지, LNG를 어디서 들여왔는지, LNG의 비율이 바뀐 것이 이산화탄소 때문인지입니다. 처음 짐작은 맞지 않은 것이 더 많았고, LNG 발전량을 가장 크게 움직인 몇 시인가는 짐작에 없었습니다.

**결정 근거** — 1-가 표(1단계 물음의 부분마다 앞 단계에서 정한 결론) · 1-가 아래 문답과 「✍️ 정할 것」 · 1-나 표 · 1-나 아래 문답과 「✍️ 정할 것」. 숫자는 앞 단계 결론 칸 · 닫는 칸과 거기서 가리키는 셀 출력(2단계 물음 2 · 4단계 2-마 · 8단계 2-다 · 2-라 · 2-사 · 9단계 2-가 · 2-나 · 3-가 ~ 3-마 · 4-나 · 10단계 1-다 · 2-가 · 2-아)에 있는 것이고, 이 노트북에서 새로 돌린 코드는 없습니다.

**남은 것**

- 물음 2 — README 「한계」. 결론 글에 적은 단서 넷(평소 범위는 한 달 LNG 발전량의 10 ~ 15%보다 작은 변화를 잡지 못한다 · 배 척 수는 발전에 쓴 LNG를 배로 바꾼 것이지 실제로 들여온 배의 수가 아니다 · 원전 · 석탄 · 전력 수요의 모델 예측은 그래디언트 부스팅이 아니라 선형회귀로 냈다 · 석탄의 비율이 2월에 이미 36%였던 것은 시기가 맞는 후보 셋으로도 설명되지 않는다)과, 앞 단계에서 「11단계 한계」로 넘긴 줄을 README 「한계」와 하나씩 맞춘다(맨 아래 「11단계 표지 — 자세한 내용」).
- 물음 3 — README 전체를 문제 → 데이터 → 방법 → 결과 → 한계로 다시 쓴다. 「결과」와 「처음 짐작과 데이터」는 이 물음에서 정한 글을 쓴다. 「1도에 얼마」(16도보다 1도 추울 때마다 LNG 발전량이 한 시각에 239 MWh 늘었다 같은 것)는 인과지도 화살표 옆에도 적는다.

**표의 숫자와 말** — 셋째 칸은 그 단계 📌 결론 칸의 「결정」 줄(1단계는 결론 칸의 문장, 9단계 물음 2의 둘째 줄은 「이 물음에서 알게 된 것」)을 옮긴 것입니다. 「평소 범위」는 평소 해에 모델 예측이 한 달 평균으로 가장 크게 틀린 크기(±2,177 MWh)입니다. 9단계 결정의 「적었던 · 많았던」은 모델 예측보다 적었던 · 많았던 것입니다(맨 위 「말 뜻」). [ ] 안은 이 표에서 풀어 적은 것입니다.

### ✍️ 다시 읽고 나서 — ① 얼마나 달랐나 (1)

9단계의해당 하는내용을 봤을 떄 보이는건 모두 다 우리가 잡았던 2177보다 작다. 즉, 다 범위안에있어. 그래서 평소 패턴과 얼마나달랐냐고 하면크게 다르지 않다. 대신 그 내부에서 크고 작은 변동은눈에 띈다(특히 6월). 이정도?

### 클로드 답변

> **짧게** — 거의 맞습니다. 고칠 곳은 하나입니다. 「크게 다르지 않다」를 「달랐다고 말할 근거가 없다(같았다는 뜻은 아니다)」로 바꿉니다.

**고칠 곳** — 평소 범위 안이라는 것은 「차이가 작았다」가 아니라 「모델이 평소 해에도 틀리던 크기와 구별되지 않는다」는 뜻입니다. 6월을 예로 들면, 실제가 모델 예측보다 한 시각 평균 1,595 MWh 적었습니다(10단계 1-다 [표 1]). 그런데 평소 해에도 모델은 한 달 평균으로 2,177까지 틀렸습니다. 그래서 이 1,595가 2026년이 정말 달라서 생긴 것인지, 모델이 원래 틀리는 만큼인지 가를 수 없습니다. 「크게 다르지 않았다」고 단정하지 않고 「달랐다고 말할 근거가 없다」로 적는 까닭입니다. 9단계 물음 2에서 「똑같다는 것은 아닌 것으로」라고 하신 것과 같은 뜻입니다.

**맞게 보신 것** — 1 ~ 6월 모두 2,177 안이라는 것, 가장 크게 달랐던 달이 6월이라는 것입니다. 하나 더하면, 달마다 방향이 있었습니다. 1 ~ 3월은 실제가 모델 예측보다 많았고(+1,096 ~ +1,446), 4 · 5월은 거의 같았고, 6월은 적었습니다(−1,595).

적으신 「그 내부의 변동」은 다음 물음에서 풀어 보겠습니다.

**②** 1-가 표에서 「주요 원인 — 2026년을 모델 예측과 다르게 만든 것」 가운데 9단계 물음 4 줄을 보시고 답해 주세요. LNG 발전량이 평소 범위 안에 머무는 동안, 그 안에서 무엇과 무엇이 서로 반대로 움직였다고 말하시겠습니까? 한 줄이면 됩니다.

### ✍️ 다시 읽고 나서 — ② 무엇이 반대로 움직였나 (1)

### 클로드 답변

> **짧게** — 맞습니다. 원전은 모델 예측보다 적었고 석탄은 많았습니다. 한쪽은 적고 한쪽은 많았던 것을 제가 「반대로」라고 쓴 것이었습니다. 고칠 곳은 「부족분을 석탄이 많이 채운거야」 하나입니다.

**고칠 곳** — 석탄이 원전 대신 만든 것은 1월부터가 아니라 2월부터이고, 갈수록 커졌습니다. 1-가 표의 10단계 물음 2 줄에 있는 「석탄과 LNG가 더 만든 양 가운데 석탄의 몫」이 1월 1%에서 5월 91%로 늘었습니다. 1월에 원전(과 그 밖의 발전원)이 덜 만든 양은 주로 LNG가 대신 만들었습니다. ①에서 말씀드린 방향, 곧 1 ~ 3월에 LNG가 모델 예측보다 많았던 것도 이것과 맞습니다. 그리고 덜 만든 양의 일부는 전력 수요가 줄어서 대신 만들 필요가 없었습니다(10단계 2-아).

기억하신 대로 비중을 비교한 곳은 10단계 물음 2가 맞습니다.

**③** 같은 10단계 물음 2 줄을 보시고 답해 주세요. 석탄이 모델 예측보다 많았던 까닭을 무엇이라고 말하시겠습니까? 한 줄이면 됩니다.

### ✍️ 다시 읽고 나서 — ③ 석탄이 많았던 까닭 (1)

답: 이거는 외적인 변화? 를 포함해서 설명해야한다고 생각했어. 그때 봤던 데이터상 4월까지는 큰 변화가없다가 5,6월에는 좀 크게 바뀌고, 비중이 늘었는데, 이건 외적인 사건 때문이라는 아이디어였던거 같어

### 클로드 답변

> **짧게** — 까닭을 모델 밖의 외부 사건에서 찾아야 한다는 것은 맞습니다. 고칠 곳은 둘입니다. 「외적인 사건 때문」은 「시기가 맞는 후보는 있지만 데이터로 가를 수 없다」로, 「4월까지는 큰 변화가 없다가」는 「2월부터 늘었다」로 바꿉니다.

**고칠 곳 1 — 「때문」** — 1-가 표의 10단계 물음 2 줄에 정하신 것은 둘입니다.

- ⓐ 값 차이(LNG와 석탄의 연료비)로는 설명되지 않는다.
- ⓑ 시기가 맞는 까닭 셋은 데이터로 가를 수 없다. 셋은 계절관리제와 3월 16일 석탄 상한 해제, 1 ~ 4월에 좁았던 값 차이, 정부의 LNG 수요 관리입니다.

적으신 답에는 ⓐ가 빠졌고, ⓑ의 「가를 수 없다」가 「때문」으로 바뀌었습니다. 시기가 맞는다는 것은 그 무렵에 같이 일어났다는 데까지입니다. 게다가 석탄은 봉쇄 같은 외부 사건(2월 28일부터)보다 앞선 1월 하순부터 이미 모델 예측보다 많았습니다(9단계 물음 4).

*(외부 자료의 날짜와 확인 정도 — 2월 28일 충돌 시작 · 3월 4일 봉쇄 선언 · 3월 16일 석탄 상한 해제는 확인(9단계 4-가 1 · 3 · 4번). 계절관리제는 해마다 12월 1일 ~ 3월 31일에 하는 제도다(9단계 4-가 5번). 정부의 LNG 수요 관리는 10단계 2-다 표 — 3월 24일 국무회의 보고는 계획, 4월 1 ~ 2일 천연가스 위기경보 「주의」는 확인.)*

**고칠 곳 2 — 「4월까지는 큰 변화가 없다가」** — 기억하신 흐름(1 ~ 4월에는 작다가 5 · 6월에 크게)은 줄어든 수요를 빼지 않고 낸 10단계 2-나 · 2-마의 비율입니다. 결론에 쓴 것은 줄어든 수요를 뺀 2-아입니다. 여기서는 석탄의 몫이 1월 1%에서 2월 36%, 3 · 4월 50%쯤, 5월 91%로, 2월부터 이미 늘었습니다.

**④** 마지막 물음입니다. 1-가 표의 「주요 원인 — 평소에 LNG를 움직이는 것」(8단계 물음 2 줄)을 보시고 답해 주세요. 평소에 LNG 발전량을 가장 크게 움직이는 것은 무엇이라고 말하시겠습니까? 한 줄이면 됩니다. 답하시면 ① ~ ④에 적으신 말을 이어서 한 문단으로 만들어 드리겠습니다.

### ✍️ 다시 읽고 나서 — ④ 평소에 LNG를 움직이는 것 (1)

답: 8단계 물음 2 결론에 적힌 것처럼, 몇시, 무슨 날, 기온, 햇빛, 몇월 순서야

### 클로드 답변

> **짧게** — ④도 맞습니다. 하나만 더하면, 이 순서는 모델에 넣은 다섯 가운데의 순서입니다. 원전 · 석탄 · 가격은 모델에 넣지 않아서 이 순서에 들어 있지 않습니다. 아래는 ① ~ ④에 적으신 말을 이어 붙인 한 문단입니다.

> 평소 해(2023 ~ 2025년)에 LNG 발전량을 가장 크게 움직인 것은 모델에 넣은 다섯 가운데 몇 시인가 → 무슨 날인가 → 기온 → 햇빛 → 몇 월인가 순서였다. 이 평소 패턴을 배운 모델이 2026년 1 ~ 6월의 날씨 · 시각 · 날 종류로 예측한 양과 실제를 비교하면, 달마다 차이가 모두 평소 범위(평소 해에 모델이 한 달 평균으로 가장 크게 틀린 크기, 한 시각 2,177 MWh) 안이었다. 그래서 평소 패턴과 달랐다고 말할 근거가 없다(같았다는 뜻은 아니다). 배로 바꾸면 평소 범위는 한 달 약 3 ~ 4척이고, 가장 크게 달랐던 6월도 약 2 ~ 3척이었다. 그 안에서 크고 작은 변동은 눈에 띄었다. 1 ~ 3월은 실제가 모델 예측보다 많았고 6월은 적었다. 원전 발전량은 1월부터 모델 예측보다 적었고, 석탄 발전량은 1월 하순부터 평일에 많았다. 원전 등이 덜 만든 양은 1월에는 주로 LNG가 대신 만들었고, 2월부터는 석탄의 몫이 늘어 5월에는 석탄과 LNG가 더 만든 양의 91%가 석탄이었다. 석탄이 많았던 까닭은 LNG와 석탄의 값 차이로는 설명되지 않는다. 계절관리제와 3월 16일 석탄 상한 해제, 1 ~ 4월에 좁았던 값 차이, 정부의 LNG 수요 관리처럼 시기가 맞는 후보는 있지만, 어느 것 때문인지는 데이터로 가를 수 없다. 석탄이 많아지기 시작한 것도 2026년의 외부 사건(2월 28일 충돌 시작)보다 앞이었다.

*(외부 자료의 날짜와 확인 정도 — 2월 28일 충돌 시작 · 3월 16일 석탄 상한 해제는 확인(9단계 4-가 1 · 4번). 계절관리제는 해마다 12월 1일 ~ 3월 31일에 하는 제도다(9단계 4-가 5번). 정부의 LNG 수요 관리는 10단계 2-다 표 — 3월 24일 국무회의 보고는 계획, 4월 1 ~ 2일 천연가스 위기경보 「주의」는 확인.)*

**바꾼 곳**

- ④를 맨 앞으로 옮겼습니다. 평소 패턴을 먼저 말해야 그다음의 비교가 읽혀서입니다.
- ①의 「크게 다르지 않다」, ②의 「부족분을 석탄이 많이 채운거야」, ③의 「외적인 사건 때문」은 앞에서 고친 대로 적었습니다.
- 배 몇 척(10단계 물음 1), 「모델에 넣은 다섯 가운데」, 「평일에」는 제가 더한 것입니다.

빼거나 고칠 곳을 말씀해 주세요.

### 클로드 답변

### ✍️ 정할 것 — README 「결과」에 넣을 글

#### 1단계 물음(2026년 LNG 발전량은 평소 패턴에서 얼마나 벗어났나, 그 차이의 주요 원인은 무엇인가)에 대한 답을 README 「결과」에 넣을 글로 적으면?

답: 모델에 넣은 다섯(몇 시인가 · 무슨 날인가 · 기온 · 햇빛 · 몇 월인가)은 모두 LNG 발전량을 움직였다. 2023 · 2024년 자료에서 크게 움직인 순서는 몇 시인가 → 무슨 날인가 → 기온 → 햇빛 → 몇 월인가였다(2023년으로 배운 모델 넷이 2024년을 맞힐 때 변수 하나의 값만 뒤섞으면 얼마나 더 틀리는지를 모델 넷에서 평균 낸 순서 — 8단계 2-다). 몇 시인가는 그다음인 무슨 날인가의 두 배쯤이고, 몇 월인가는 몇 시인가의 10분의 1쯤이었다. 기온과 햇빛은 크기가 비슷해서 모델에 따라 순서가 바뀌고, 햇빛은 해가 떠 있는 시각에만 LNG 발전량을 움직인다. 기온과 일사가 바뀔 때 얼마나 움직였는지 보면, 16도보다 1도 추울 때마다 LNG 발전량이 한 시각에 239 MWh, 24도보다 1도 더울 때마다 661 MWh 늘었고, 낮에 일사가 1 MJ/m² 많을 때마다 2,329 MWh 줄었다(2023 · 2024년으로 배운 모델 넷의 평균 — 8단계 2-사. 일사는 단위가 기온과 달라 크기를 서로 비교하지 않는다).

이 다섯을 넣고 2023 ~ 2025년을 배운 그래디언트 부스팅으로 2026년 1 ~ 6월의 날씨 · 시각 · 날 종류에서 LNG 발전량을 예측해 실제와 비교했다(5월 2 ~ 17일은 받은 자료의 값이 잘못돼 있어 뺐다). 그 차이는 달마다 모두 평소 범위 안이었다. 평소 범위는 그래디언트 부스팅이 그 전 해까지를 배워 2024 · 2025년을 예측했을 때 한 달 평균으로 가장 크게 틀린 크기다(한 시각 평균 2,177 MWh). 날로 나눠 봐도 2024 · 2025년에 가장 크게 틀린 날보다 크게 틀린 날은 없었고, 낮(10 ~ 15시)과 밤(0 ~ 5시)으로 나눠 봐도 2024 · 2025년에 가장 크게 틀린 달보다 크게 틀린 달은 없었다(9단계 물음 3). 그래서 2026년 1 ~ 6월 LNG 발전량이 평소 패턴과 달랐다고 말할 근거가 없다(같았다는 뜻은 아니다). 다만 이 기준은 한 달 LNG 발전량의 10 ~ 15%보다 큰 변화만 잡는다. 발전에 쓴 LNG를 배로 바꾸면(30일 기준) 평소 범위는 약 3 ~ 4척이고, 가장 크게 달랐던 6월도 약 2 ~ 3척이었다(실제로 들여온 배의 수가 아니다).

그 안에서 크고 작은 변동은 눈에 띄었다. LNG 발전량은 1 ~ 3월에 실제가 모델 예측보다 많았고, 4 · 5월에는 차이가 작았고, 6월에는 적었다. 원전 · 석탄 · 전력 수요의 모델 예측은 그래디언트 부스팅이 아니라 선형회귀로 냈다. 같은 2023 ~ 2025년을 같은 날씨 · 시각 · 날 종류로 발전원마다 따로 배운 것이다(9단계 2-나). 이 선형회귀로 낸 LNG 발전량의 실제 − 모델 예측은 여섯 달 모두 그래디언트 부스팅과 + · − 가 같았고, 둘의 차이는 가장 큰 5월도 한 시각 평균 231 MWh였다. 원전 발전량은 2026년 시작부터 모델 예측보다 적었다(까닭 후보는 2026년이 되기 전에 멈춘 원전 넷이다 — 멈춘 날은 확인, 9단계 4-가). 석탄 발전량은 1월 하순(1월 22일이 든 주)부터 평일에 모델 예측보다 많았고, 달로 보면 2월부터 많았다. 전력 수요(전력시장에서 거래된 양 — 전기를 덜 썼는지, 전력시장 밖의 태양광 발전이 늘었는지는 가를 수 없다)는 5월을 뺀 달에 모델 예측보다 적었다. 원전 발전량이 적었던 것은 LNG 발전량을 늘리는 쪽이고, 석탄 발전량이 많았던 것과 전력 수요가 적었던 것은 줄이는 쪽이라, 이 둘이 같은 달에 상쇄되어 LNG 발전량의 차이는 평소 범위 안이었다(9단계 물음 4). 석탄과 LNG의 발전량이 각각 모델 예측보다 많았던 양을 합치면, 그 가운데 석탄이 차지한 비율은 1월 1%(거의 다 LNG), 2월 36%, 3월 52%, 4월 50%, 5월 91%였다(10단계 2-아 — 선형회귀로 낸 값. 6월은 LNG가 모델 예측보다 적어 내지 않았다).

6월에 LNG 발전량이 모델 예측보다 적었던 것을 낮(10 ~ 15시)과 밤(0 ~ 5시)으로 나눠 보면, 낮에는 전력 수요가 모델 예측보다 적었던 양이, 밤에는 석탄 발전량이 모델 예측보다 많았던 양이 가장 컸다(9단계 4-라). 날 종류로 나누면 토 · 일만 2024 · 2025년의 토 · 일보다 크게 틀렸는데(실제가 모델 예측보다 많았다), 비교할 해가 둘뿐인 약한 기준이라 판정은 바꾸지 않았다(9단계 물음 3). 9단계 물음 4에서는 이것을 석탄 발전량이 평일에만 모델 예측보다 많았던 것으로 설명했다.

석탄 발전량이 모델 예측보다 많았던 까닭은 이 프로젝트에서 찾지 못했다. LNG와 석탄의 연료비단가 차이로는 설명되지 않는다. 연료비단가(EPSIS — 발전소가 전기 1 kWh를 만드는 데 드는 연료비로, 배출권 비용도 들어 있는 것으로 봤다(판단). LNG는 두 달쯤 전에 들여온 연료의 값이라 시장 가격이 늦게 반영된다)는 1 ~ 6월 내내 석탄이 LNG보다 낮았는데도 석탄의 비율은 달마다 크게 달랐고, 1 ~ 4월에는 LNG가 석탄보다 비싼 정도가 평소 해보다 오히려 작았는데도 석탄 발전량이 모델 예측보다 많았다(10단계 물음 2). 석탄의 비율이 커진 것과 시기가 맞는 것은 셋이다(외부 자료의 날짜와 확인 정도는 10단계 2-다 표). 첫째, 겨울(12 ~ 3월) 미세먼지 계절관리제(석탄 발전소의 가동을 멈추거나 출력을 설비의 80%로 낮추게 하는 제도 — 석탄 발전량을 줄이는 쪽이다)와, 그 출력 상한을 3월 16일에 해제한 것(확인). 둘째, 1 ~ 4월에 평소 해보다 작았던 연료비단가 차이가 5월에 평소 해와 같아진 것. 셋째, 3월부터의 정부의 LNG 수요 관리(3월 24일 국무회의에 보고된 원전 재가동 · 석탄 발전 탄력 운영은 계획, 4월 1 ~ 2일에 천연가스 위기경보를 「주의」로 올린 것은 확인). 셋은 시기가 맞는 후보일 뿐이고, 무엇이 까닭인지는 이 데이터로 가를 수 없었다. 석탄의 비율이 2월에 이미 36%였던 것은 셋으로도 설명되지 않는다 — 2월에 걸리는 것은 계절관리제와 연료비단가 차이인데, 계절관리제는 석탄 발전량을 줄이는 쪽이고, 연료비단가 차이는 비율이 1%였던 1월과 같았다(LNG가 석탄보다 1 kWh에 1월 47.0원 · 2월 47.5원 비쌌다. 판단 — 10단계 물음 2 결론의 한계 11). 석탄 발전량이 모델 예측보다 많아지기 시작한 때(1월 하순)도 호르무즈 사태(2월 28일 충돌 시작 — 확인, 9단계 4-가)보다 앞이었다(9단계 물음 4).

LNG가 들어오고 쌓이고 쓰이는 양 가운데 이 프로젝트가 본 것은 LNG 발전량 하나다. 재고와 들여온 양은 보지 못해서, 2026년에 LNG가 모자랐는지는 말할 수 없다.

- **① 판정** — 2023 → 2025년에는 짐작대로 LNG의 몫이 늘고 석탄의 몫이 줄었지만, 2026년 1 ~ 6월에는 그렇게 움직이지 않았다(원전 등이 덜 만든 양을 2월부터 석탄이 점점 더 많이 대신 만들었다). 탄소가 적어서인지는 이 프로젝트로 볼 수 없다.
- **README에 넣을 줄** — 처음 짐작 여섯 줄은 대체로 빗나갔다. 맞은 것은 추위 하나였고, 계절 영향은 반대였다. 가격이 정한다는 짐작은 2026년 1 ~ 6월에 맞지 않았고, 탄소와 지정학(카타르 LNG) 짐작은 일부만 맞거나 이 자료로 볼 수 없었다. 호르무즈 사태가 나면 LNG에 바로 문제가 나타날 줄 알았는데, LNG 발전량에는 바로 나타나지 않았다. LNG를 가장 크게 움직인 몇 시인가는 여섯 줄에 없었다(데이터를 보고도 넣지 않았다).

**README에 넣을 줄(고친 것)**

> 처음에 짐작한 여섯 가지는 대체로 맞지 않았다. 맞은 것은 「추우면 난방 수요가 늘어 LNG 발전량이 는다」 하나였다. 「LNG는 발전에 쓰이니 계절에 따라 크게 달라지지 않을 것이다」는 반대였다. LNG 발전량은 계절에 따라 크게 달랐다. 「값이 싼 연료를 더 많이 쓴다」는 2026년 1 ~ 6월에 맞지 않았다. 석탄이 늘 LNG보다 쌌는데도, 석탄과 LNG가 더 만든 전기 가운데 석탄이 만든 비율은 1월 1%에서 5월 91%로 달마다 크게 달랐다. 「LNG는 같은 열을 낼 때 석탄보다 이산화탄소를 적게 내니 LNG 비율이 커질 것이다」는 2023 → 2025년에는 맞았지만 2026년 1 ~ 6월에는 맞지 않았다. 「호르무즈 사태로 카타르 LNG가 줄면 석탄과 원전이 더 발전할 것이다」는 석탄만 맞았고, 원전 발전량은 오히려 모델 예측보다 적었다. LNG를 어디서 들여왔는지는 이 자료로 알 수 없었다. 호르무즈 사태가 나면 LNG 발전량에 바로 변화가 나타날 줄 알았는데, 사태 뒤에도 LNG 발전량은 평소 범위 안이었다. 그리고 LNG 발전량을 가장 크게 움직인 「몇 시인가」는 여섯 가지에 없었다(데이터를 보고도 넣지 않았다).

### ✍️ 정할 것 — 짐작 ①의 판정과 README 「처음 짐작과 데이터」

#### 1. 짐작 ①의 판정을 1-나 표 ① 줄에 넣을 만큼 짧게 적으면?

답: 세 해의 전력거래량에서 LNG가 차지한 비율은 2023년 28.2%, 2024년 29.8%, 2025년 29.1%로 2023년보다는 커졌지만 해마다 늘지는 않았다(2단계 물음 2). 2026년 1 ~ 6월은 그 비율을 내지 않고 모델 예측과 비교했다. 석탄과 LNG가 모델 예측보다 많았던 양 가운데 1월은 거의 다 LNG였고(짐작과 같은 쪽), 2월부터는 석탄의 비율이 커졌다(짐작과 다른 쪽, 5월 91%). 이산화탄소 때문인지는 볼 자료가 없었다.

#### 2. 짐작 ①의 판정을 자세히 적으면?

답: 2023 ~ 2025년에 LNG가 차지한 비율은 해마다 늘지는 않았다. EPSIS의 한 해 전력거래량으로 세 해를 이어 보면 LNG의 비율은 2023년 28.2%, 2024년 29.8%, 2025년 29.1%로 2024년에 늘었다가 2025년에 줄었고, 석탄은 32.9%, 29.4%, 30.1%로 그 반대였다(2단계 물음 2). 2023년과 2025년 두 해만 놓고 보면 LNG의 비율은 늘었다. 8월을 뺀 한 해 전력거래량에서 LNG는 28.2%에서 29.2%로 1.0%p 늘었고, 석탄은 32.6%에서 29.2%로 3.4%p 줄었으며, 원자력은 31.7%에서 32.8%로, 태양광은 1.8%에서 2.3%로 늘었다(4단계 2-마 — 연간 파일에 연료가 다 들어 있는 2023 · 2025년 두 해만 비교했다). LNG의 비율이 늘거나 준 까닭이 이산화탄소 때문인지는 이 표들로 가를 수 없다.

2026년 1 ~ 6월은 전력거래량에서 LNG가 차지한 비율을 따로 내지 않았고, 모델 예측과 비교했다. 원전 발전량은 2026년 시작부터 모델 예측보다 적었다. 석탄과 LNG의 발전량이 각각 모델 예측보다 많았던 양을 합치면, 그 가운데 석탄이 차지한 비율은 1월 1%, 2월 36%, 3월 52%, 4월 50%, 5월 91%였다(10단계 2-아 — 선형회귀로 낸 값). 1월은 모델 예측보다 많았던 양이 거의 다 LNG라 짐작과 같은 쪽이었고, 2월부터는 석탄의 비율이 커져 짐작과 다른 쪽이었다(판단). 석탄 발전량은 1월 하순부터 평일에 모델 예측보다 많았다. 배출권거래제 4차 계획기간(2026 ~ 2030년)에 발전 부문이 돈을 내고 사야 하는 배출권 비율을 10%에서 15%로 늘릴 계획이었는데도(계획 — 9단계 4-가), 석탄 발전량은 모델 예측보다 많았다.

석탄 발전량이 모델 예측보다 많았던 까닭은 이 프로젝트에서 찾지 못했다. 까닭을 좁히며 확인한 것은 셋이다. 첫째, LNG와 석탄의 연료비단가 차이로는 설명되지 않는다. 연료비단가는 1 ~ 6월 내내 석탄이 LNG보다 낮았는데도 석탄의 비율은 달마다 크게 달랐고, 1 ~ 4월에는 LNG가 석탄보다 비싼 정도가 평소 해보다 오히려 작았는데도 석탄 발전량이 모델 예측보다 많았다(10단계 물음 2). 이 연료비단가에는 배출권 비용도 들어 있는 것으로 봤다(판단 — 따로 계산하지 않았다). 그렇다면 배출권 비용까지 넣고도 석탄이 1 ~ 6월 내내 LNG보다 쌌다. 둘째, 석탄 발전량이 모델 예측보다 많아지기 시작한 때(1월 22일이 든 주)는 호르무즈 사태(2월 28일 충돌 시작 — 확인, 9단계 4-가)보다 앞이라, 그 사태가 시작의 까닭일 수는 없다(9단계 물음 4). 셋째, 3월 16일에 석탄 발전 상한을 해제한 뒤 석탄 발전량이 평소 해보다 늘었다는 근거가 없다. 날마다 석탄 발전량이 가장 많았던 한 시각의 값을 해제 전 14일(3월 2 ~ 15일)과 뒤 14일(3월 16 ~ 29일)로 나눠 평균 내면 뒤 14일이 1,449 MWh 낮았는데, 평소 해(2023 ~ 2025년)에도 같은 날짜 앞뒤로 759 ~ 2,553 MWh 낮아져서 2026년은 그 범위 안이었다(9단계 4-나 — 날씨를 맞추지 않은 비교다. 날씨를 맞춘 4-가에서도 3월 16일 뒤에 석탄 발전량이 늘지 않았다).

석탄의 비율이 커진 것과 시기가 맞는 것은 셋이다(외부 자료의 날짜와 확인 정도는 10단계 2-다 표). 첫째, 겨울(12 ~ 3월) 미세먼지 계절관리제(석탄 발전소의 가동을 멈추거나 출력을 설비의 80%로 낮추게 하는 제도 — 석탄 발전량을 줄이는 쪽이다)와, 그 출력 상한을 3월 16일에 해제한 것(확인). 둘째, 1 ~ 4월에 평소 해보다 작았던 연료비단가 차이가 5월에 평소 해와 같아진 것. 셋째, 3월부터의 정부의 LNG 수요 관리(3월 24일 국무회의에 보고된 원전 재가동 · 석탄 발전 탄력 운영은 계획, 4월 1 ~ 2일에 천연가스 위기경보를 「주의」로 올린 것은 확인). 셋은 시기가 맞는 후보일 뿐이고, 무엇이 까닭인지는 이 데이터로 가를 수 없었다. 석탄의 비율이 2월에 이미 36%였던 것은 셋으로도 설명되지 않는다 — 2월에 걸리는 것은 계절관리제와 연료비단가 차이인데, 계절관리제는 석탄 발전량을 줄이는 쪽이고, 연료비단가 차이는 비율이 1%였던 1월과 같았다(LNG가 석탄보다 1 kWh에 1월 47.0원 · 2월 47.5원 비쌌다. 판단 — 10단계 물음 2 결론의 한계 11).

LNG가 같은 열을 낼 때 석탄보다 이산화탄소를 적게 낸다는 점 때문에 비율이 바뀌었는지는, 이 프로젝트에 그것을 볼 자료가 없어서 말할 수 없다.

#### 3. README 「처음 짐작과 데이터」에 넣을 글은?

답: 처음에 짐작한 여섯 가지는 맞지 않은 것이 더 많았고, 반대로 움직인 것도 있었다. 맞은 것이 하나(②), 반대였던 것이 하나(③), 「어느 연료가 더 싼가」로는 2026년 1 ~ 6월에 맞지 않은 것이 둘(④ · ⑥), 일부만 맞은 것이 둘(① · ⑤)이다. 특히 호르무즈 사태가 나면 LNG 발전량이 바로 줄 것이라고 생각했는데 그러지 않았다. 여섯 가지는 데이터를 일부 본 뒤(09-18)에 적은 것이라, 데이터를 보기 전에 세운 가설은 아니다.

- ① LNG는 같은 열을 낼 때 석탄보다 이산화탄소를 적게 내니, 재생에너지로 바뀌어 가는 동안 LNG를 쓰는 비중이 조금씩 커질 것이다 — 세 해의 전력거래량에서 LNG가 차지한 비율은 2023년 28.2%, 2024년 29.8%, 2025년 29.1%로 2023년보다는 커졌지만, 2024년에 늘었다가 2025년에 줄어 해마다 늘지는 않았다(석탄은 32.9%, 29.4%, 30.1%로 그 반대였다). 2026년 1 ~ 6월에는 석탄과 LNG가 모델 예측보다 많았던 양 가운데 석탄의 비율이 2월부터 커져(5월 91%, 선형회귀로 낸 값) 짐작과 다른 쪽이었다(1월은 거의 다 LNG). 이산화탄소 때문인지는 볼 자료가 없었다.
- ② 추우면 난방 수요가 늘어 LNG를 많이 쓸 것이다 — 맞았다. 16도보다 1도 추울 때마다 LNG 발전량이 한 시각에 239 MWh 늘었다. 왜 느는지(전기 난방으로 전력 수요가 늘어서인지, 열병합 발전소가 겨울에 난방열과 전기를 같이 만들어서인지)는 보지 못했다.
- ③ 발전용 LNG라면 그 비율이 계절에 크게 영향받는다고 예측할 수도 없을 것 같다(겨울에 도시가스로 많이 쓰면 그만큼 발전용 LNG는 줄 것) — 반대였다. 발전용 LNG도 계절에 따라 크게 달랐고, 겨울에는 줄지 않고 많았다(2023 · 2024년 실제 LNG 발전량의 한 시각 평균이 1월 20,648 MWh, 10월 15,225 MWh). 그 차이는 대부분 달마다 다른 기온 · 햇빛 때문이었다 — 날씨는 그대로 두고 달만 1월 · 10월로 바꿔 넣으면 그래디언트 부스팅이 예측한 평균이 17,672 · 17,121 MWh로 551 MWh 차이였다(8단계 2-라). 도시가스로 쓴 LNG 때문에 발전용 LNG가 줄었는지는 도시가스 자료가 없어 모른다.
- ④ 연료별 효율과 가격, 경제성이 무엇을 쓸지 정한다 · ⑥ 결국 상대적인 가격과 에너지원마다의 특성이 LNG 발전량을 정한다 — 「어느 연료가 더 싼가」로는 2026년 1 ~ 6월에 맞지 않았다. 연료비단가는 1 ~ 6월 내내 석탄이 LNG보다 낮았는데도, 석탄과 LNG가 모델 예측보다 많았던 양 가운데 석탄의 비율은 1월 1%에서 5월 91%로 달마다 크게 달랐다. 다만 연료비단가 차이가 1 ~ 4월에 평소 해보다 작았다가 5월에 평소 해와 같아진 것은 석탄의 비율이 커진 때와 시기가 맞아 까닭 후보로 남았다(판단).
- ⑤ 호르무즈 사태로 카타르에서 가져오는 LNG가 줄었다면 다른 곳에서 가져와야 했을 텐데 어디서 가져왔을까 — 석탄이나, 원자력 발전량을 유연하게 늘릴 수 있다면 원자력에서 가져올 것이다 — 어디서 가져왔는지는 알 수 없었다(자료에는 발전에 쓴 LNG만 있고 들여온 LNG는 없다). 석탄 발전량이 모델 예측보다 많았던 것은 짐작과 같은 쪽이었지만, 그 크기는 평소 해에도 있던 정도였고, 사태 전인 1월 하순부터 많았고, 사태 때문인지는 가를 수 없었다. 원전 발전량은 오히려 모델 예측보다 적었고, LNG 발전량은 사태 뒤에도 평소 범위 안이었다(3월은 오히려 모델 예측보다 많았다).

여섯 가지에 없던 것도 있다. LNG 발전량을 가장 크게 움직인 몇 시인가와 햇빛 · 더운 쪽 기온은 09-13에 데이터나 이야기로 보고도(햇빛은 클로드가 꺼낸 이야기였다) 여섯 가지에 넣지 않았다. 무슨 날인가 · 몇 월인가는 여섯 가지에 없었지만, 적을 때 데이터로 이미 알고 있었다.

### 클로드 답변

## 📌 물음 2 결론

**왜 생긴 물음인가** — README 「한계」는 단계마다 그때그때 적어 둔 서른네 줄이라, 앞 단계에서 11단계 한계로 적기로 하고 넘긴 것이 다 들어 있는지 맞춰 본 적이 없었습니다. 넘어온 한계를 「한계」와 하나씩 맞춰, 빠진 것은 더하고 「결과」 글과 어긋나는 말은 고치는 물음입니다.

### 결정 — 앞 단계에서 넘어온 한계는 모두 남긴다. 「한계」에 없던 경계 문단과 열두 줄을 넣고, 이미 있던 줄은 「결과」 글과 같은 말로 고쳤다. 「한계」는 경계 문단과 마흔여섯 줄이다

**같이 정한 것**

1. 경계 문단(LNG가 들어오고 쌓이고 쓰이는 양 가운데 본 것은 LNG 발전량 하나라는 것 · LNG 값에서 LNG 발전량으로 바로 가는 화살표를 지운 까닭 · 시스템 다이내믹스로 한 일)을 「한계」 맨 앞에 둔다.
2. 2-가의 열두 줄(뺀 기간 셋과 남긴 22시각 · 2026년 7월 · 발전소마다의 자료 · 아주 추운 시각 · 서로 다른 모델로 낸 것 · 평소 범위로 잡지 못하는 변화 · 날 종류 기준 · 9 ~ 11월에 몰린 날 · 선형회귀 0 ~ 9시 · 모델을 고른 기준 · 핵심인자 순서 · 3월 16일 앞뒤 비교)과 「달로 나눠 보면 기준선보다 더 틀린 달도 있었다」 한 줄을 끝에 더한다.
3. 이미 있던 줄은 「예측이 실제보다」를 「실제가 모델 예측보다」로 바꾸고, 어느 해 · 어느 모델의 값인지 밝히고, 계획으로 적힌 채 남은 말을 한 일로 고치고, 근거 칸과 다른 말을 고치고, 출처 · 확인 정도 · 판단 표시를 붙인다(2-가 아래 답변의 표 둘 — 클로드가 정함).
4. 「한계」를 묶고 줄이는 것은 물음 3에서 한다(클로드 안).

**이 물음에서 알게 된 것** — 넣을 것의 목록은 앞 단계에서 정리돼 있었지만, 그것이 README에 실제로 들어가 있는지는 따로 맞춰야 했습니다. 이미 있던 줄도 단계마다 그때의 말로 적혀 있어서, 어느 해 · 어느 모델의 값인지가 빠지거나 계획으로 적힌 채 남은 곳이 있었습니다. 「한계」는 결과를 어디까지 믿어도 되는지를 말하는 곳이라 「결과」 글과 같은 말로 적어야 합니다.

**결정 근거** — 2-가(넘어온 한계 가운데 「한계」에 없는 것과 더할 글) · 2-가 아래 「✍️ 내 답변」(창주 님 「너가 적합하게 내가 여기 전체에 적힌 한계를 제대로 적었다면뭐 들어가는거고. 그러니까 적절하게 살펴서 한계로 적히면 적합한거 잘 적으면 될것 같어.」) · 그 아래 답변의 표 둘 · 「✍️ 다시 읽고 나서」(「좋아 이정도면」).

**남은 것**

위 「이 물음을 왜 하나」와 표를 읽고, 창주 님 말로 한 줄 적어 주세요.

### 클로드 답변

물음 1에서 검토한 것은 README 「결과」와 「처음 짐작과 데이터」 글입니다. 「한계」 서른네 줄은 단계마다 그때그때 적은 것이라, 앞 단계에서 「11단계에서 한계로 적는다」며 넘긴 것이 다 들어 있는지는 아직 맞춰 본 적이 없습니다(물음 1에서는 결과 글과 겹치는 두 줄만 말을 맞췄습니다).

순서는 이렇습니다.

1. 제가 — 넘어온 한계와 물음 1의 단서 넷을 「한계」 줄과 맞춰, 이미 있는 것 · 없는 것(더할 글까지) · 말을 고칠 것을 표 하나로 11단계 노트북 `11_결론`의 2-가에 넣습니다.

3. 제가 — README 「한계」에 넣고 물음 2를 닫습니다.

## 2-가 — 앞 단계에서 넘어온 한계 가운데 README 「한계」에 없는 것과 더할 글

**이 셀이 답하는 물음** — 「이게 어차피 정리된거 아닌가?」(창주 님) — 정리된 목록이 README 「한계」에 실제로 들어가 있나, 없는 것은 무엇을 더하나.

**어디서 나온 셀인가** — 물음 2를 열 때 클로드가 계획한 셀입니다(클로드 제안). 맨 아래 「11단계 표지 — 자세한 내용」에서 물음 2로 넘어온 다섯 줄, 📌 물음 1 결론 「남은 것」의 단서 넷, 3 ~ 10단계 닫는 칸에서 「11단계 한계」로 넘긴 줄을 README 「한계」 서른네 줄과 하나씩 맞췄습니다. 3단계 닫는 칸 1번에 적힌 말은 「이 문단이 README 한계와 11단계 결론에 그대로 들어갑니다」입니다. 말을 고칠 줄(쓰지 않기로 한 말 · 「결과」 글과 어긋나는 말)은 이 셀에 답하신 뒤 2-나로 드립니다 — 한 칸에 몰면 읽기 어려워서입니다(클로드 판단).

**이 칸의 뜻** — 맞춰 보고 무엇을 더할지는 클로드가 정했습니다. 글은 앞 단계 닫는 칸 · 결론 칸에 적힌 말을 그대로 쓰고, 쓰지 않기로 한 말만 바꿨습니다. 빼거나 고칠 줄은 아래 「✍️ 내 답변」에 적으시면 그대로 따릅니다.

**이미 「한계」에 있어 그대로 두는 것** — 넣어 보지 않은 다섯(부산 기상 · 일사를 달마다 · 월 × 시간대 · 불쾌지수 · 태양광 설비를 아는 열) · 늘어난 태양광 설비 · 30도 넘는 시각과 부산 기온 · 습한 날과 불쾌지수 · 원전 정비 일정과 석탄을 모델이 모르는 것 · 크게 틀린 날이 며칠씩 이어진 것 · 주말에 든 공휴일 · 토요일 · 고른 모델의 2025년 성적 · 모델 넷을 평균 내는 방법을 해 보지 않은 것 · 잔차가 이웃한 시각끼리 이어지는 것 · 자가용 · PPA 태양광 · 기온이 서울 한 지점인 것 · 계절 반복이 두 번인 것 · 배 척 수(단서 넷의 둘째) · 석탄이 많았던 까닭과 2월의 석탄 비율(단서 넷의 넷째) · 전력 수요 · 그 밖의 발전원 · 배출권 비용 · 석탄 발전소 정비 · 1단계 짐작 ⑤의 물음(카타르 LNG가 줄면 어디서 가져왔을까) · LNG 값이 두 달쯤 늦게 따라오는 것 · 원전 설비용량.

**넣을 곳** — 경계 문단은 「한계」 제목 바로 아래에, 나머지 열둘은 「한계」 끝에 덧붙입니다. 「한계」를 묶고 줄이는 것은 물음 3에서 합니다(클로드 안). 지금 「한계」의 「LNG가 들어오고 쌓이고 쓰이는 양 가운데 이 프로젝트가 본 것은 LNG 발전량 하나다」 줄은 경계 문단과 겹쳐서 2-나에서 정리합니다.

---

### README 「한계」에 더할 글

**경계 문단 — 이 프로젝트가 본 것과 보지 못한 것.** LNG가 들어오고 쌓이고 쓰이는 양 가운데 이 프로젝트가 본 것은 LNG 발전량 하나다. 2026년에 모델 예측과 달랐던 원전 · 석탄 · 전력 수요와 그 까닭 후보(원전은 2026년이 되기 전에 멈춘 원전 넷, 석탄은 계절관리제와 3월 16일 상한 해제 · 1 ~ 4월에 좁았던 LNG와 석탄의 연료비단가 차이 · 정부의 LNG 수요 관리 — 「결과」)는 모두 모델 밖에 있고, 재고와 들여온 양은 보지 못해 LNG가 모자랐는지는 말할 수 없다(10단계 물음 3). 재고가 버틴 뒤 7월 이후에 나타났을 영향도 말할 수 없다(10단계 3-나). 인과지도에서 LNG 값이 LNG 발전량으로 바로 이어지는 화살표는 7단계에서 지웠다. 한국 전력시장은 연료비가 싼 발전소의 전기부터 쓰고 LNG는 그러고 남는 만큼 발전해서, 석탄이 LNG보다 싼 동안은 LNG 값이 올라도 쓰는 순서가 바뀌지 않기 때문이다(10단계 2-가 — 2023 ~ 2026년 1 ~ 6월의 모든 달에 유연탄이 LNG보다 쌌다). 그래서 LNG 값은 석탄 발전 상한과 전기 요금을 거쳐서만 LNG 발전량으로 이어진다. 보는 기간의 사정 때문이 아니라 시장의 구조 때문이다. 이 프로젝트에서 시스템 다이내믹스로 한 일은 인과지도로 모델이 보는 것과 보지 못하는 것을, 저량 · 유량 그림으로 자료가 본 것과 보지 못한 것을 갈라 그린 것이고, 시뮬레이션은 하지 않았다.

1. **받은 자료에서 뺀 기간이 셋이다(3단계).** 2024년 2월 29일은 API에 그날이 아예 없다. 2024년 8월 1 ~ 31일은 API 기록이 잘못 들어 있어 학습에서 뺐다 — 다시 요청해도 같은 값이 왔고 EPSIS 월별 표에는 그 달이 정상으로 들어 있어, 실제 발전이 아니라 기록의 문제다. 그래서 2023 · 2024년으로 배운 모델은 8월을 2023년 한 해치만 배웠고, 해끼리 평균이나 합계를 비교할 때는 세 해에서 8월을 똑같이 빼고 비교한다(4단계). 2026년 5월 2 ~ 17일도 받은 자료의 값이 잘못돼 있어 2026년 비교에서 뺐다. 남긴 것도 하나 있다 — 상자그림에서 벗어난 22시각(2023년 3월 일요일 낮 아홉 · 1 · 4 · 12월 평일 열셋)은 값이 잘못 들어온 것이 아니라 조건이 극단적이었던 시각이라 학습에 남겼다.
2. **2026년 7월은 비교에서 뺐다(9단계 물음 1).** 받은 7월 자료가 EPSIS 월 합계와 0.99% 어긋났다 — 2023 ~ 2025년에 가장 크게 어긋난 달(0.24%)의 네 배쯤이다. 그래서 2026년 비교는 6월 30일까지라, 호르무즈 봉쇄(3월 4일) 뒤로는 넉 달만 봤다(9단계 닫는 칸 ③).
3. **발전소마다의 자료가 없다.** 원전 정비 일정은 한수원 API를 네 가지 방식으로 불러 봤지만 날짜와 상관없이 최근 엿새만 줘서 쓰지 못했다(2a단계, 10-27). 그래서 원전 발전량이 달마다 왜 움직였는지(정비인지)는 판단으로만 말한다. 9단계 4-다에서 원전 발전량을 4월 4일 · 4월 11일 앞뒤로 비교한 것도 원전 전체를 합한 값이라, 같은 주에 다른 원전이 정비에 들어가거나 마친 것이 섞인다. 발전기마다의 운전 기록도 없어서 2026년 주말에 석탄 발전기를 어떻게 운전했는지는 보지 못했다 — 주말 석탄 발전량이 모델 예측보다 적었던 것이 6 ~ 15시였다는 데까지 봤다(9단계 4-바).
4. **아주 추운 시각은 자료를 더해도 드물다(4단계, 09-13).** 서울 한 지점에서 영하 12도 이하인 시각은 2023년 64시간 · 2024년 10시간 · 2025년 2시간이다. 아주 추운 날의 예측은 이 적은 시각으로 배운 값이다(2025년 영하 5도 이하 시각은 잘 맞혔다 — 7단계 3-바).
5. **「얼마나 달랐나」와 「무엇이 달랐나」는 서로 다른 모델로 냈다(9단계 물음 1 · 2-나).** LNG 발전량이 평소 범위 안인지는 그래디언트 부스팅으로 봤고, 원전 · 석탄 · 전력 수요의 「실제 − 모델 예측」과 석탄의 비율은 선형회귀로 냈다. 선형회귀는 차이를 수요 · 원전 · 석탄 · 그 밖의 넷으로 나누면 그 합이 차이와 꼭 맞아서 나눠 읽을 수 있고, 그래디언트 부스팅은 그렇게 나눌 수 없다.
6. **평소 범위로는 작은 변화를 잡지 못한다.** 달 단위 기준(평소 범위, 한 시각 평균 ±2,177 MWh)은 한 달 LNG 발전량의 10 ~ 15%보다 큰 변화만 잡는다(9단계 물음 2 — 2026년 1 ~ 6월 LNG의 한 시각 평균 14,987 ~ 22,710 MWh에 비교한 값). 날 단위 기준(2024 · 2025년에 가장 크게 틀린 날 6,271 · 위 5%가 시작하는 값 4,210 — 그날 시각마다 틀린 양의 평균, 한 시각 MWh)으로는 몇 주에 걸친 작은 변화를 잡지 못한다. 9단계 3-나의 7일 평균 선으로 읽기만 했고, 주 단위 기준은 두지 않았다(9단계 물음 3). 그래서 「달랐다고 말할 근거가 없다」는 「같았다」는 뜻이 아니다.
7. **날 종류로 나눈 기준은 약하다(9단계 3-마).** 2024 · 2025년 두 해 가운데 더 크게 틀린 해의 크기를 기준으로 삼아서, 날 종류 일곱 줄 가운데 2.3줄쯤은 우연으로도 넘는다(줄끼리 서로 상관없다고 친 어림 — 판단). 2026년 토 · 일이 그 기준을 넘은 것은 이 약점과 같이 읽는다.
8. **2024 · 2025년에 크게 틀린 날은 9 ~ 11월에 몰려 있었다(9단계 3-가).** 그래디언트 부스팅이 가장 크게 틀린 열 날 가운데 여섯이 9 ~ 11월이었다(날 수대로라면 2.6날). 2026년 비교는 1 ~ 6월이라 그 까닭은 보지 않았다.
9. **선형회귀는 2025년 0 ~ 9시에 기준선보다 더 틀렸다(7단계 3-다).** 0 ~ 5시에 한 시각 평균 2,887 MWh를 틀려 기준선(2,492)보다 컸고, 6 ~ 9시도 기준선보다 컸다. 기온이 LNG를 바꾸는 크기를 시간대마다 따로 배운 선형회귀도 0 ~ 5시에는 2,594로 기준선보다 컸다. 2025년을 본 뒤라 고치지 않았다.
10. **8단계에서 모델을 고른 기준 하나는 결과를 보기 전에 정했지만, 클로드가 그 기준에 동의한 것은 결과를 본 뒤였다(8단계 물음 1).** 「같은 달이 2024 · 2025년 두 해 모두 기준선보다 더 틀린 모델을 뺀다」로 덜 엄격하게 정한 것은 8단계 1-나 결과를 보기 전이었다. 한 달이라도 기준선보다 더 틀린 달을 다 세는 쪽으로 했어도 그래디언트 부스팅이 남았다.
11. **핵심인자의 순서는 미리 기준으로 정한 셀이 아니라 다른 셀에서 가져왔다(8단계 물음 2).** 기준으로 정한 8단계 2-나(그 변수만 흔히 있는 값으로 바꿔 넣었을 때 모델이 맞힌 LNG가 움직인 폭)는 햇빛만 낮 10 ~ 15시로 폭을 구해서, 다섯 변수를 같은 시각 범위로 비교한 순서가 아니었다. 그래서 순서는 2-다(그 변수의 값만 다른 시각의 값으로 바꿔 놓았을 때 모델이 더 틀린 양)에서 가져왔는데, 2-다도 새벽 2시에 한낮의 햇빛 값이 붙는 것처럼 실제로는 없는 조합을 만든다. 어느 셀로 적을지는 두 셀의 순서를 본 뒤에 정했고, 두 셀 모두 2023 · 2024년 자료만 썼다.
12. **석탄 발전 상한 해제(3월 16일) 앞뒤 비교는 날씨를 맞추지 않았다(9단계 4-나).** 모델을 쓰지 않은 실제 발전량끼리의 비교라 해마다 다른 날씨가 섞여 있다. 날씨를 맞춘 9단계 4-가(주마다 석탄 발전량의 「실제 − 모델 예측」)에서도 2026년은 3월 16일 뒤에 오르지 않았다.

# 물음 3 — README 전체를 처음 보는 사람이 읽을 판으로

**적는 물음입니다 — README를 처음 보는 사람이 위에서부터 읽고 「무엇을 왜 어떻게 했고 무엇이 나왔나」를 알 수 있게 다시 쓰고, 공개할 판에 무엇을 담을지 정하면 끝납니다. 창주 님 말대로 이것은 나중의 창주 님에게도 해당해서, README만 읽고 「3단계에서 뭐 했죠?」에 답할 수 있어야 끝납니다.**

**지금 어디**

- 데이터 — 9단계까지 시간별 LNG · 기온 · 햇빛으로 모델 예측을 만들어 2026년 1 ~ 6월과 비교했고, 10단계에서 그 차이를 해석했습니다.
- 모델에 들어가 있는 것 — 몇 시인가 · 몇 월인가 · 무슨 날인가 · 기온 · 햇빛. 원전 · 석탄 · 가격은 넣지 않았습니다.
- 물음 1에서 정한 것 — 「2026년 1 ~ 6월 LNG 발전량은 달마다 평소 범위 안이라 평소 패턴과 달랐다고 말할 근거가 없다.」 README 「결과」와 「처음 짐작과 데이터」에 그 글을 넣었습니다.
- 물음 2에서 정한 것 — 앞 단계에서 넘어온 한계를 모두 README 「한계」에 넣었습니다(경계 문단과 마흔여섯 줄). 묶고 줄이는 것은 이 물음에서 합니다.

**이 물음을 왜 하나** — 지금 README는 대부분이 진행 기록(「지금 어디」의 옛 판 · 걸어온 자취)이라, 처음 보는 사람은 무엇을 했고 무엇이 나왔는지 찾기 어렵습니다. 어느 프로젝트에서나 이 물음이 하는 일은 같습니다 — 프로젝트를 하며 쌓은 기록을, 과정을 모르는 사람이 읽을 수 있는 글 하나로 바꾸는 일입니다.

| 앞에서 정한 것 | 이 물음이 정할 것 | 뒤에서 할 것 |
|---|---|---|
| README 차례는 문제 → 데이터 → 방법 → 결과 → 한계다(지침 11단계 칸). 「결과」 · 「처음 짐작과 데이터」 글은 물음 1에서, 「한계」 글은 물음 2에서 정했다. 그림은 10단계 인과지도와 저량 · 유량 그림을 둘 다 넣는다(10단계 물음 3). 이 저장소는 비공개로 두고, 다듬은 판은 같은 깃허브 아이디 아래 새 저장소에 올린다(09-20 · 10-29) | 지금 흩어져 있거나 앞 단계의 말로 남은 「문제 · 데이터 · 방법」을 처음 보는 사람이 읽을 글로 새로 쓴다. 그림 둘을 넣고 깃허브 첫 화면에서 보이는지 본다. 진행 기록은 `docs/`로 옮겨 첫 설계부터 한 표로 잇고, 「한계」 마흔여섯 줄은 묶어 줄인다(클로드 안). 3분 안에 설명할 수 있게 한다. 공개할 판에 무엇을 담을지 고른다 | 물음 4 — 다음 프로젝트 전에 공부할 것. 단계를 끝낼 때 — 4 · 5단계 개념 표의 빈 「내 말로」 여덟 칸, 노트북 형식 맞추기, 커밋. 다 끝난 뒤 — 핵심만 뽑아 다듬은 판을 새 저장소에 올린다 |

**적을 곳** — 먼저 바로 아래 「✍️ 이 물음은 무엇을 정하는 물음입니까?」(채팅에서 적으신 말을 옮겼습니다), 그다음 3-가 아래 「✍️ 내 답변」입니다. 셀마다 왜 생겼는지는 아래 🗂️ 칸에 있습니다.

**어디서 나온 물음인가** — 지침 11단계 칸(README — 인과지도와 저량 · 유량 그림 한 장씩, 경계 한 문단) · README 11단계 줄(3분 안에 설명할 수 있게, 남에게 보여 줄 파일을 고른다) · 📌 물음 1 결론과 📌 물음 2 결론의 「남은 것」 · 맨 아래 「11단계 표지 — 자세한 내용」에서 물음 3으로 넘어온 줄.

답: 결국 지금까지 한 전체적인 흐름을 말 그대로 요약 정리하는 부분이라고 생각해. 결론과 한계가 앞 물음에서 만들어 졌으니까 이제 처음 보는 사람에게 이 2달의 내용을 전달하는 걸 다듬는거지, 이건 나에게도 해당하는 말이야. 긴 시간동안 했는데 당연히 지금 나에게 3단계에서 뭐 했죠? 이러면 헷갈리잖어?? 그러니까 전체저그으로 흐름을 보여주고, 핵심적인 프로젝트의 결과와 한계 핵석의 방향과 뭐 그런것들이 쭉 적히는거라고 판단했어

### 클로드 답변

> **짧게** — 네, 맞습니다. 결론과 한계는 물음 1 · 2에서 정했고, 물음 3은 두 달의 흐름을 처음 보는 사람이 위에서부터 읽고 따라올 수 있게 다듬는 물음입니다. 「나에게도 해당하는 말」은 물음 3이 끝나는 기준으로 머리 칸 첫 줄에 적었습니다 — README만 읽고 「3단계에서 뭐 했죠?」에 답할 수 있으면 끝입니다.

적으신 말에서 README에 담을 것은 넷입니다 — 전체 흐름 · 핵심 결과 · 한계 · 해석의 방향. 해석의 방향은 결과를 어떻게 읽는지로 읽었습니다. 예를 들어 평소 범위 안이면 「달랐다고 말할 근거가 없다」로 읽고 「같았다」로 읽지 않는다, 시기가 맞는 후보는 까닭이 아니다 같은 것입니다. 지금은 README 「말 뜻」 표와 「무엇을」 칸의 「말할 수 있는 것과 없는 것」에 흩어져 있어서, 「결과」 맨 앞에 모으려고 합니다(아래 3-가).

순서는 이렇습니다.

1. 제가 — 아래 3-가에 README 새 판의 차례를 드립니다.
2. 창주 님 — 차례를 보시고 빼거나 더하거나 순서를 바꿀 칸을 3-가 아래에 적으십니다.
3. 제가 — 차례대로 칸마다 글을 하나씩 드리면, 읽히는지 보시고 정하십니다. 공개할 판에 무엇을 담을지는 README를 다 쓴 뒤 고릅니다.

📘 4과목(빅데이터 결과 해석) — 분석 결과를 받는 사람이 이해하고 쓸 수 있게 정리하고 그림으로 보이는 일(분석결과 시각화 · 활용)에 해당합니다.

## 3-나 — 전체 흐름: 단계마다 한 일과 정한 것

**이 셀이 답하는 물음** — README만 읽고 「3단계에서 뭐 했죠?」에 답할 수 있게, 11단계마다 무엇을 했고 무엇을 정했나(위 창주 님 한 줄에서 — 전체 흐름을 보여 준다).

**어디서 나온 셀인가** — 3-가에서 정한 차례의 둘째 칸입니다. 글의 길이와 말은 채팅에서 3단계 줄로 맞췄습니다. 비유와 줄인 말을 빼고(창주 님 「여기에는 즉 결론 칸에는 비유적이거나, 너만 아는 함축적으로 붙인 표현은 빼줘. 딱딱해도 알아먹을 수 있게.」), 확인 과정까지 적은 판에 「좀 너무 상세해서 흐름을 잃을 거 같은데?」라고 하셔서 단계마다 한 일과 정한 것만 두세 문장으로 두었고, 그렇게 고친 3단계 줄에 「좋아」라고 하셨습니다. 그래서 3-가 아래 답변에 예로 든 「까닭을 붙여 뺐다」는 쓰지 않습니다.

**표 보는 법** — 한 줄이 한 단계입니다. 「한 일과 정한 것」은 README에 그대로 들어갈 글이고, 「README에서 자세히」는 그 단계의 자료와 숫자를 README의 어느 칸에 둘지입니다. 표 아래 「단계를 오간 것」은 순서대로 하지 않은 곳입니다. 글은 앞 단계 📌 결론 칸에 적힌 것만 썼습니다.

| 단계 (끝낸 날) | 한 일과 정한 것 | README에서 자세히 |
|---|---|---|
| 1 과제 정의 (09-18) | 답할 물음을 「2026년 LNG 발전량은 평소 패턴에서 얼마나 벗어났는가」로 정하고, 그 차이의 주요 원인을 살펴보는 것까지 물음에 넣었다. 예측 대상은 전력시장에서 거래된 시간별 LNG 발전량(MWh)이다. LNG 발전량을 늘리거나 줄일 것 같은 원인을 짐작으로 여섯 줄 적었다. | 문제 |
| 2 데이터 수집과 구성 보기 (09-19) | 전력거래소(시간별 발전량) · 기상청(서울의 시간별 기온 · 습도 · 일사) · 전력통계정보시스템(EPSIS, 월별 발전량) · 공휴일 자료를 받아, 몇 줄 몇 칸인지와 빠지거나 겹친 시각이 있는지 봤다. 2023 ~ 2025년 LNG 발전량의 월 합계는 값이 잘못된 기간이 든 달을 빼면 EPSIS와 맞았다. | 데이터 |
| 3 정제 (09-19) | 전력거래소에서 받은 시간별 발전량 가운데 세 기간(2024년 2월 29일 · 2024년 8월 · 2026년 5월 2 ~ 17일)은 값이 비어 있거나 잘못 들어 있었다. 실제 발전이 그랬던 것이 아니라 받은 자료가 잘못된 것이어서, 이 기간은 모델이 배우는 데에도 2026년과 비교하는 데에도 쓰지 않았다. 2023년에 이상치로 나온 22시각은 값이 잘못된 것이 아니라 조건이 극단적이었던 시각이라 남겼다. | 데이터 |
| 4 탐색적 분석 (09-22) | 무엇이 LNG 발전량을 늘리거나 줄이는지 그림과 상관분석(두 값이 함께 늘고 주는 정도를 −1 ~ 1의 상관계수로 나타내는 것)으로 봤다. LNG 발전량과 상관계수가 가장 큰 것은 몇 시인가와 주말인가였다. 기온은 18 ~ 20도쯤에서 LNG 발전량이 가장 적고, 그보다 춥거나 더울수록 늘었다. 낮에는 태양광 발전량이 많을수록 LNG 발전량이 줄었다. 석탄 발전량은 LNG 발전량보다 먼저 바뀌지 않고 같은 시각에 함께 정해져서, 입력에 넣지 않기로 한 것을 그대로 두었다. | 방법 |
| 5 변환 · 파생변수 (09-30) | 4단계에서 남긴 입력 후보(몇 시인가 · 요일 · 공휴일 · 기온 · 태양광 같은 것)를 모델에 넣을 열로 만들었다. 기온은 그대로 넣는 열과 함께, 16도보다 몇 도 추운지와 24도보다 몇 도 더운지를 나타내는 열(선형회귀에 넣을 열)을 만들었다. 공휴일과 명절은 요일과 함께 날 종류(화 ~ 금 · 토 · 일 · 공휴일 · 명절)로 넣었다. 태양광은 2026년 자료에 발전량이 따로 없어서 일사(햇빛 세기)를 넣어 봤다. 2023년으로 배워 2024년을 맞힐 때 일사를 넣은 쪽이 태양광 발전량을 넣은 쪽보다 더 잘 맞혀서, 일사로 정했다. | 방법 |
| 6 분석계획 (10-05) | 원전 발전량은 넣어 보니 「원전 발전량이 줄면 LNG 발전량도 준다」로 배웠다. 원전이 덜 발전하면 LNG 같은 다른 발전원이 더 발전해야 하므로, 그 반대로 배운 것이다(2024년에 석탄이 줄면서 원전과 LNG가 같이 늘어서 생긴 일이다). 2024년도 더 못 맞혀서 입력에서 뺐다. 습도는 넣어도 2024년을 맞히는 정도가 거의 그대로여서 뺐다. 그다음 2023 · 2024년으로 배워 2025년에 한 번 채점하고, 2026년 1 ~ 6월과 비교하기로 했다. 모델이 이것보다 잘 맞혀야 쓸 이유가 있는 기준선(배운 해에서 같은 달 · 같은 시각 · 같은 날 종류인 시각들의 평균)과 시험할 모델 넷을 정했다. 모델이 보는 것과 보지 못하는 것은 인과지도로 그렸다. | 방법 · 결과(인과지도) |
| 7 모델링과 검정 (10-09) | 2025년으로 한 번 채점해, 모델 넷 모두 기준선보다 잘 맞힌 것을 확인했다. 낮에는 LNG 발전량을 실제보다 많게 예측했고(태양광 설비가 늘어서), 30도 넘는 시각과 토요일에 크게 틀렸다. 달마다 모델이 LNG를 더 많게 또는 더 적게 예측한 것은 12달 가운데 10달이 원전 · 석탄 발전량이 모델 예측과 달랐던 것과 같은 쪽이었다. 채점을 본 뒤 모델을 고치면 그 점수가 실제보다 좋게 나와서, 모델은 고치지 않고 틀린 곳을 한계로 적었다. | 방법 · 한계 |
| 8 모델 확정과 핵심인자 (10-17) | 결과를 보기 전에 정한 고르는 방법 둘(두 해(2024 · 2025년) 모두 더 잘 맞히고 나눠 봐도 꾸준히 나은 모델 · 2025년 결정계수가 가장 높은 모델)이 모두 그래디언트 부스팅을 골랐다. 모델에 넣은 다섯이 모두 LNG 발전량을 움직였고, 2023 · 2024년 자료에서 몇 시인가 → 무슨 날인가 → 기온 → 햇빛 → 몇 월인가 순서로 크게 움직였다. 1단계 짐작 여섯 줄과 맞춰 보니 맞은 것은 추위 하나였다. | 방법 · 결과(처음 짐작과 데이터) |
| 9 모델 적용 (10-27) | 그래디언트 부스팅을 2023 ~ 2025년으로 다시 배워 2026년 1 ~ 6월의 LNG 발전량을 예측하고 실제와 비교했다. 차이는 달마다 평소 범위(2024 · 2025년에 모델 예측이 한 달 평균으로 가장 크게 틀린 크기 — 2026년 LNG를 보기 전에 정한 기준) 안이라, 평소 패턴과 달랐다고 말할 근거가 없었다. 그 안에서 원전 발전량이 1월부터 모델 예측보다 적었던 것과 석탄 발전량이 평일에 많았던 것이 상쇄됐다. | 결과 |
| 10 결과 해석 (10-29) | 9단계의 차이를 LNG 배 몇 척으로 바꿔 봐도 달마다 평소 범위 안이라는 판정은 같았다. 석탄 발전량이 많았던 까닭을 찾아봤지만, LNG와 석탄의 연료비단가 차이로는 설명되지 않았고 까닭은 찾지 못했다. 이 프로젝트가 본 것(LNG 발전량 하나)과 보지 못한 것(재고 · 들여온 양 같은 것)을 인과지도와 저량 · 유량 그림으로 그렸다. | 결과 · 한계 |
| 11 결론 (10-30 ~) | 1단계 물음에 대한 답을 적고, 앞 단계에서 넘어온 한계를 모으고, 처음 보는 사람이 읽을 수 있게 이 README를 다시 썼다. | 결과 · 한계 |

## 3-다 — README 「문제」 칸의 글

**이 셀이 답하는 물음** — 처음 보는 사람이 무엇을 물었고, 무엇을 예측했고, 무엇은 하지 않았는지 알 수 있게 README 「문제」 칸을 어떻게 쓰나(3-가에서 정한 차례의 셋째 칸).

**어디서 나온 셀인가** — 3-가에서 정한 차례대로 쓰는 칸입니다. 글은 채팅에서 먼저 보여 드리고 맞췄습니다. 「맞히는 값」은 「맞히는 값? 이게 좀 에이아이 틱한 좀 어색한 표현같고,」라고 하셔서 「예측 대상(종속변수 y)」으로 바꿨고, 달마다 숫자를 어떻게 냈는지는 「결과」 맨 앞(결과를 읽는 법)으로 옮겼습니다. 「이래도 적합하나? 이게 이해는 앞문장만 있어서 잘 되는데 , 서렴ㅇ이 부족하진 않어>」에는, 무엇을 예측하는지는 앞 문장으로 충분하고 빠진 것은 고른 까닭이라 1단계 결론의 까닭 한 문장을 더했습니다. 글은 1단계 📌 결론 칸 · 지침 「몸통」(가격 · 석탄 · 총 전력거래량을 넣지 않는 까닭) · 6단계 📌 물음 1 결론(원전)에 적힌 것만 썼습니다.

> **왜** — 수치적인 데이터로 살펴보는 게 목적이다. 느낌이 아니라. *(2026-09-18에 내가 쓴 문장)*
>
> **물음** — 2026년 LNG 발전량은 평소 패턴에서 얼마나 벗어났는가. 그 차이의 주요 원인은 무엇인가.
>
> - 평소 패턴 — 평소 해(2023 ~ 2025년)에 날씨 · 시각 · 날 종류에 따라 LNG 발전량이 나온 패턴이다. 이 패턴을 배운 모델이 2026년의 날씨 · 시각 · 날 종류를 보고 예측한 양을 「모델 예측」이라고 부른다.
> - 벗어났는가 — 2026년 실제 LNG 발전량이 모델 예측과 얼마나 달랐는가다. 날씨 · 시각 · 날 종류가 같을 때 평소 패턴과 비교한 차이다.
>
> **예측 대상(종속변수 y)** — 전력시장에서 거래된 시간별 LNG 발전량(MWh). 모델은 한 시각마다 LNG 발전량을 예측한다. 이것을 고른 것은, 모델의 예측과 비교해 볼 실제 값이 전력거래소 자료에 시간마다 기록돼 있어서다(처음 세운 설계는 비교해 볼 실제 값이 데이터에 없어서 그만뒀다 — 전체 흐름의 「단계를 오간 것」).
>
> **하지 않은 것**
>
> - 가격은 예측하지 않고 입력에도 넣지 않았다. 2026년에 달라진 것이 가격이라, 넣으면 평소와 얼마나 달랐는지가 0에 가까워진다.
> - 원전 · 석탄 발전량과 총 전력거래량도 입력에 넣지 않았다. 총 전력거래량을 넣으면 「LNG = 총 전력거래량 − 나머지 발전량」이라는 뺄셈이 되어 모델이 필요 없어진다. 석탄은 LNG와 같은 시각에 함께 정해지고, 2026년에 달라진 것 자체다. 원전은 넣으면 반대로 배웠다(전체 흐름 6단계). 이 셋은 결과를 해석할 때 차이 옆에 놓았다.
> - 예측 대상은 전국 발전용 LNG이고, 특정 회사가 사들이는 LNG의 양이 아니다.

## 3-라 — README 「데이터」 칸의 글

**이 셀이 답하는 물음** — 무엇을 어디서 받아 어디에 썼고, 어떤 기간을 왜 뺐나(3-가에서 정한 차례의 넷째 칸).

**어디서 나온 셀인가** — 3-가에서 정한 차례대로 쓰는 칸입니다. 전체 흐름(3-나)의 2 · 3단계 줄에서 뺀 확인 과정(같은 날짜를 다시 요청해 본 것 · EPSIS 월 합계와 맞춰 본 것)을 여기에 두었습니다. 채팅에서 먼저 보여 드렸습니다. 글은 2 · 3단계 📌 결론 칸, 2단계에서 받은 자료의 기간, 9단계 📌 물음 1 결론(2026년 7 · 8월)에 적힌 것만 썼습니다.

**README에 넣을 글**

> **2. 데이터 (2 · 3단계)**
>
> **받은 자료**
>
> | 자료 | 어디서 받았나 | 기간 | 어디에 썼나 |
> |---|---|---|---|
> | 연료마다의 시간별 발전량 | 전력거래소 — 2023 · 2025년은 공공데이터포털의 연간 파일, 2024 · 2026년은 공공데이터포털 API(2024년 연간 파일에는 LNG가 없다) | 2023년 1월 ~ 2026년 8월 22일 | LNG 발전량은 예측 대상. 원전 · 석탄 · 태양광 같은 다른 발전원은 결과를 해석할 때 |
> | 서울의 시간별 날씨 | 기상청 종관기상관측(ASOS) 서울(108) — 공공데이터포털 API | 2023년 1월 ~ 2026년 8월 | 기온 · 일사는 모델 입력(습도는 넣어 보고 뺐다) |
> | 연료마다의 월별 발전량 | 전력통계정보시스템(EPSIS) | 2002년 1월 ~ 2026년 7월 | 받은 시간별 발전량의 월 합계가 맞는지 확인 · 해마다 LNG가 차지한 비율 |
> | 공휴일 | `holidays` 패키지로 만든 표 | 2023 ~ 2026년 | 모델 입력(날 종류) |
> | 연료마다의 연료비단가 | EPSIS 「연료비용」 | 2023년 1월 ~ 2026년 9월 | 석탄 발전량이 많았던 까닭을 찾을 때(10단계) |
>
> **받지 않은 해** — 2020 ~ 2022년은 받지 않았다. 2020년은 코로나로 전력거래량이 줄었고, 2021년은 LNG 발전량이, 2022년은 원전 발전량이 크게 늘어서 평소 해로 삼기 어려웠다(EPSIS로 확인). 그래서 모델이 배운 해는 두세 해뿐이다(「한계」).
>
> **뺀 기간과 그 까닭(3단계)**
>
> - 2024년 2월 29일 — API에 그날 자료가 없었다.
> - 2024년 8월 — API가 연료 다섯 가지만 주고, LNG 값이 다른 날의 49분의 1쯤이었다. 모델이 배우는 데서 뺐다. 그래서 8월은 다른 달보다 적은 해로 배웠다.
> - 2026년 5월 2 ~ 17일 — LNG가 빠져 있거나 값이 잘못 들어 있었다. 2026년과 비교하는 데서 뺐다(5월은 1일과 18 ~ 31일만 비교했다).
> - 2024년 8월과 2026년 5월은 같은 날짜를 다시 요청해도 같은 값이 왔고, EPSIS 월 합계에는 정상으로 들어 있었다. 실제 발전이 아니라 받은 자료의 문제다.
> - 2026년 7 · 8월 — 받은 7월 값이 EPSIS 월 합계와 0.99% 어긋나고(평소 해에 가장 크게 어긋난 달은 0.24%), 8월은 EPSIS에 아직 없어 확인할 수 없어서 비교에 넣지 않았다(9단계).
> - 남긴 것 — 2023년 상자그림에서 이상치로 나온 22시각은 대부분 한겨울 평일 아침이나 3월 일요일 한낮이었다. 값이 잘못 들어온 것이 아니라 조건이 극단적이었던 시각이라 남겼고, 빼도 기온과 LNG 발전량의 상관계수가 거의 그대로였다.

## 3-마 — README 「방법」 칸의 글

**이 셀이 답하는 물음** — 모델에 무엇을 넣고 무엇을 뺐고, 어느 해로 배우고 채점하고 비교했고, 무엇을 기준으로 정했나(3-가에서 정한 차례의 다섯째 칸).

**어디서 나온 셀인가** — 3-가에서 정한 차례대로 쓰는 칸입니다. 지금 README의 「결정에 쓴 기준」 표는 처음 보는 사람이 읽게 말을 고쳐 이 칸 끝에 두었습니다(9단계 줄도 이 표에 둡니다). 채팅에서 먼저 보여 드렸습니다. 글은 4 ~ 8단계 📌 결론 칸 · 9단계 📌 물음 1 결론 · 지침 「규칙」(누수)에 적힌 것만 썼습니다.

**README에 넣을 글**

> **3. 방법 (4 ~ 8단계)**
>
> **모델에 넣은 것** — 그래디언트 부스팅에 일곱 열을 넣었다.
>
> - 몇 시인가(0 ~ 23시) · 몇 월인가
> - 무슨 날인가 — 요일 · 공휴일인가 · 명절인가
> - 기온 · 일사(햇빛 세기) — 둘 다 서울 한 지점의 시간별 값
>
> **넣지 않은 것과 까닭**
>
> - 원전 발전량 — 넣으면 「원전 발전량이 줄면 LNG 발전량도 준다」로 반대로 배웠다(6단계).
> - 석탄 발전량 · 총 전력거래량 · 가격 — 「문제」의 「하지 않은 것」에 적었다.
> - 태양광 — 발전량은 2026년 자료에 따로 없고, 설비용량은 2025년 값이 모두 배운 해의 범위 밖이라 일사로 대신했다(3 · 5단계).
> - 습도 · 불쾌지수 — 넣어도 2024년을 맞히는 정도가 아래 기준만큼 나아지지 않았다(6단계).
> - 풍력 · 바이오매스 · 연료전지 — 시간마다 오르내리는 폭이 LNG 발전량의 5% 안팎이거나 거의 일정해서 후보에서 뺐다(4단계).
>
> **배우고, 채점하고, 비교한 해** — 해를 시간 순서대로 나눴다(앞의 해로 배우고 뒤의 해를 맞혔다).
>
> - 넣을 열을 고르는 동안(5 · 6단계)은 2023년으로 배워 2024년을 맞혀 봤다. 이때 2025 · 2026년 LNG 발전량은 쓰지 않았다. 답을 미리 보고 고르면 채점이 처음 보는 해의 점수가 아니게 되기 때문이다(그 전에 본 것은 「한계」에 적었다).
> - 고른 열로 2023 · 2024년을 배워 2025년에 한 번 채점했다. 채점을 본 뒤에는 모델을 고치지 않았다.
> - 같은 모델을 2023 ~ 2025년으로 다시 배워 2026년 1 ~ 6월을 예측하고 실제와 비교했다.
>
> **기준선과 모델 넷** — 기준선은 배운 해에서 같은 달 · 같은 시각 · 같은 날 종류인 시각들의 평균을 그대로 예측으로 쓰는 것이다. 모델은 이것보다 잘 맞혀야 쓸 이유가 있다. 모델은 넷(선형회귀 · 기온 × 시간대 선형회귀 · 랜덤포레스트 · 그래디언트 부스팅)을 시험했고, 2025년 채점에서 넷 모두 기준선보다 결정계수가 높았다(기준선 0.668, 모델 0.701 ~ 0.755). 결정계수는 예측이 실제의 오르내림을 얼마나 설명하는지 나타내는 값으로, 1에 가까울수록 잘 맞힌 것이다. 고른 모델은 그래디언트 부스팅이다(아래 표의 8단계 줄).
>
> **결정에 쓴 기준** — 단계마다 「무엇과 비교해서, 얼마나 나아야(달라야) 의미가 있다고 할까」를 결과를 보기 전에 적어 두고 그대로 정했다.
>
> | 언제 | 무엇을 정하려고 | 기준(결과를 보기 전에 적은 것) | 그래서 |
> |---|---|---|---|
> | 5 · 6단계 | 이 열을 넣을까 | 2023년으로 배워 2024년을 맞혔을 때 결정계수가 0.002 넘게 오르나(기온을 나누는 기준 온도만 바꿔도 이만큼은 달라져서 정한 크기) · 배운 방향이 4단계에서 본 것과 반대가 아닌가 | 원전 · 습도 같은 열을 뺐다 |
> | 6 · 7단계 | 모델을 쓸 이유가 있나 | 기준선보다 결정계수가 높은가 — 6단계는 2024년, 7단계는 2025년으로 한 번 | 모델 넷 모두 기준선보다 나았다 |
> | 8단계 | 넷 가운데 어느 모델로 갈까 | 같은 달이 두 해(2024 · 2025년) 모두 기준선보다 더 틀리지 않았나 · 반대로 배운 것이 없나 · 두 해 모두 더 잘 맞히고, 달 · 시간대 · 기온 · 날 종류로 나눈 63묶음의 절반 넘게 덜 틀리나(2025년 결정계수가 가장 높은 모델도 나란히 봤다) | 그래디언트 부스팅 — 두 방법이 같은 모델을 골랐다 |
> | 8단계 | 어느 변수가 크게 작용했나 | 그 변수를 같은 크기만큼 바꿨을 때 LNG 발전량 예측이 가장 크게 바뀌는가 | 결론의 순서(몇 시인가 → 무슨 날인가 → 기온 → 햇빛 → 몇 월인가)는 결과를 본 뒤 변수의 값을 뒤섞어 보는 방법으로 골랐다(「한계」) |
> | 9단계 | 2026년이 평소와 달랐나 | 1 ~ 6월 가운데 「실제 − 모델 예측」의 한 달 평균 크기가 평소 범위(2024 · 2025년에 모델 예측이 한 달 평균으로 가장 크게 틀린 크기, 한 시각 2,177 MWh)를 넘는 달이 두 달 이상인가 | 넘은 달이 없어서, 달랐다고 말할 근거가 없다 |

## 3-바 — README 「결과」 칸 맨 앞의 「결과를 읽는 법」

**이 셀이 답하는 물음** — 결과를 어떻게 읽어야 하나(창주 님 한 줄의 해석의 방향). 「결과」 칸의 결과 글과 「처음 짐작과 데이터」는 물음 1에서 정했고, 그 앞에 이 글을 둡니다.

**어디서 나온 셀인가** — 3-가에서 「결과」 칸 맨 앞에 두기로 한 글입니다. 지금 README의 「말 뜻」 표와 「무엇을」 칸의 「말할 수 있는 것과 없는 것」을 모았고, 3-다에서 옮겨 온 달마다 숫자 내는 법(1월 744시각의 예)을 더했습니다. 채팅에서 먼저 보여 드렸고, 넷째 줄은 「이것만 좀 안읽히고,」라고 하셔서 예를 붙여 길게 고쳤다가, 「아니 오히려 이게 다른 것들과 다른 형식으로 적혀있는데 다른건 좀 매끈하게 적히는데 이건 좀 주절주절해. 기존거 설명좀 덧붙이거나 어려운 말대신 쉬운말로 하면 안돼나ㅣ?」라고 하셔서 다른 줄과 같은 꼴로 두고 쉬운 말로만 바꿨습니다. 글은 README 「말 뜻」 · 「결과」 글 · 9단계 📌 물음 1 결론 · 지침 「규칙」(외부 자료의 확인 정도)에 적힌 것만 썼습니다.

**README에 넣을 글**

> **결과를 읽는 법**
>
> | 말 | 뜻 |
> |---|---|
> | 실제 | 그해(주로 2026년)에 실제로 발전한 양. 전력거래소 자료 그대로다 |
> | 모델 예측 | 2023 ~ 2025년을 배운 모델이 그때의 날씨 · 시각 · 날 종류를 보고 예측한 양 |
> | 실제 − 모델 예측 | 발전량 표의 숫자. + 면 실제가 모델 예측보다 많았고, − 면 적었다 |
> | 평소 해 | 2023 · 2024 · 2025년. 「평소」는 이 세 해만 가리킨다 |
> | 평소 범위 | 평소 해에 모델 예측이 한 달 평균으로 가장 크게 틀린 크기(한 시각 ±2,177 MWh). 모델이 예측해 본 2024 · 2025년으로 냈다 |
> | 세 해 평균 | 2023 ~ 2025년 같은 달 값의 평균. 모델 없이 계산한 것으로, 연료비단가를 비교할 때만 쓴다 |
>
> - 달마다의 차이는 그 달 모든 시각의 「실제 − 모델 예측」을 평균 낸 것이다. 예를 들어 1월의 차이는 1월 744시각마다 실제 − 모델 예측을 구해 평균 낸 값이다. 그 크기(+ · −를 뗀 값)가 평소 범위보다 크면 그달은 평소 범위 밖이다.
> - 평소 범위 안이면 「달랐다고 말할 근거가 없다」로 읽는다. 「같았다」는 뜻이 아니다. 이 기준은 한 달 LNG 발전량의 10 ~ 15%보다 작은 변화는 잡지 못한다.
> - 모델 밖의 일(호르무즈 사태 · 가격 · 석탄 발전 상한 해제 · 원전 발전량이 줄어든 것)이 차이의 얼마를 만들었는지는 구하지 않았다. 같은 달 원전 · 석탄 · 전력 수요가 모델 예측과 얼마나 달랐는지를 LNG의 차이 옆에 놓고 읽었다.
> - 발전량이 달라진 때와 같은 때에 일어난 일(정책이나 사건)은 까닭의 「후보」로만 적었다. 같은 때에 일어났다고 그 일 때문이라고 할 수는 없고, 후보가 여럿이면 이 자료로는 무엇 때문인지 구별할 수 없었다.
> - 괄호 안의 표시 — 「확인」은 일어난 뒤에 쓴 기사 둘 이상이나 정부 발표 원문이 있는 것, 「계획」은 하겠다는 발표만 있는 것, 「판단」은 자료로 확인하지 않고 해석한 것이다.

## 3-사 — README에 넣을 그림 둘과 그 설명 줄

**이 셀이 답하는 물음** — README의 그림 둘을 어디에, 어떤 설명과 함께 넣나(창주 님 한 줄의 「프로젝트를 처음 본 사람이 이해할 수 있도록」).

**어디서 나온 셀인가** — 10단계 물음 3에서 README에 그림 둘(인과지도 · 저량 · 유량 그림)을 넣기로 했고, 8단계 물음 2에서 「1도에 얼마」(16도보다 1도 추울 때마다 LNG 발전량이 한 시각에 239 MWh 는다 같은 숫자)를 인과지도 화살표 옆에 적기로 했습니다. 적는 곳을 두 가지로 할 수 있어서 채팅에서 여쭈었습니다.

| | 가 — 그림 파일에 적는다 | 나 — README의 그림 바로 아래 줄에 적는다 |
|---|---|---|
| 하는 일 | 인과지도 파일에서 추운 정도 · 더운 정도 · 일사 옆에 숫자를 넣고, Vensim에서 다시 캡처 | 그림은 그대로 두고, README에서 그림 바로 아래 한 줄에 세 숫자를 적는다 |
| 대가 | 다시 캡처를 한 번 해야 하고, 글자가 화살표와 겹치면 더 해야 한다 | 그림만 따로 떼어 보면 숫자가 없다 |

클로드 판단은 나였습니다(같은 숫자가 바로 위 결과 글에도 있고, README를 읽는 사람은 그림과 그 아래 줄을 같이 본다). 설명 줄은 처음에 「나머지는 모델 밖이다」 · 「자료로 본 것은 발전에 쓴 LNG 하나다」로 썼다가, 「모델 밖이다 대신, 뭐 모델에서 고려하지 않기로 결정한 것들이다. 뭐 이렇게 적히는게 좋을 것 같고.. 발전에쓴 lng하나다?? 이거 뭔말인지 모르겠는데 너무 맥락 함축저깅야」라고 하셔서 고쳤습니다. 굵은 글자가 아닌 것에는 일부러 뺀 것도 있지만 자료가 없어 넣을 수 없었던 것도 있어서 둘로 나눠 적었고, 저량 · 유량 그림은 칸 이름으로 풀어 적었습니다.

**그림 둘** — 이 노트북에서는 아래처럼 보이고, README에서는 `figures/` 폴더의 같은 파일을 가리킵니다.

![인과지도 — 10단계](../figures/인과지도_10단계.png)

![저량 · 유량 그림 — 10단계](../figures/저량유량_10단계.png)

**README에 넣을 글**

> **「결과」 끝 — 인과지도 아래**
>
> 인과지도 — LNG 발전량을 늘리거나 줄이는 것과 그 사이의 화살표다. 굵은 글자는 모델에 입력으로 넣은 변수다(LNG 발전량은 예측 대상). 나머지는 모델에 넣지 않은 것이다. 일부러 뺀 것(석탄 발전 · 원전 발전 · 전력 수요 · LNG 가격 · 습도 같은 것)과, 자료가 없어 넣을 수 없었던 것(사회 활동 · 시장 밖 태양광 · 정부의 LNG 수요 관리 같은 것)이 섞여 있다. 화살표 ① ~ ⑧에는 10단계에서 근거(봤다 · 확인 · 계획 · 판단 · 2026년엔 아님 · 못 봄)를 달았다(그림 아래쪽 줄).
>
> 그림에서 모델이 보는 변수 셋이 LNG 발전량을 바꾼 크기(한 시각, 2023 · 2024년으로 배운 모델 넷의 평균) — 16도보다 1도 추울 때마다 +239 MWh · 24도보다 1도 더울 때마다 +661 MWh · 낮에 일사가 1 MJ/m² 많을 때마다 −2,329 MWh

> **「한계」 맨 앞 — 저량 · 유량 그림 아래**
>
> 저량 · 유량 그림 — LNG가 나라 안으로 들어오고(들여온 LNG) 저장 탱크에 쌓였다가(LNG 재고) 발전소와 도시가스로 나가는(발전에 쓴 LNG · 도시가스로 쓴 LNG) 흐름을 그린 것이다. 이 가운데 이 프로젝트가 자료로 가진 것은 발전소가 LNG로 만든 전기의 양(LNG 발전량)뿐이다. 들여온 LNG · LNG 재고 · 도시가스로 쓴 LNG는 자료가 없어 보지 못했다.

## 3-아 — README 「한계」를 넷으로 묶은 글

**이 셀이 답하는 물음** — 「한계」 마흔여섯 줄을 처음 보는 사람이 읽을 수 있게 묶고 줄이면 어떻게 되나(창주 님 한 줄의 「프로젝트를 처음 본 사람이 이해할 수 있도록」).

**어디서 나온 셀인가** — 물음 2에서 앞 단계에서 넘어온 한계를 모두 남기기로 했고, 묶고 줄이는 것은 물음 3에서 하기로 했습니다. 클로드가 묶는 방법(넷으로 묶고, 줄마다 붙은 진행 기록은 빼고 두세 문장으로 줄인다)을 정해 채팅에서 보여 드렸고, 「이게 우리가 본것들의 전체라면 뭐 좋아.」라고 하셨습니다. 그래서 지금 「한계」의 마흔여섯 줄이 어디로 갔는지를 먼저 표로 둡니다.

| 묶은 판의 묶음 | 지금 「한계」의 줄 | 줄 수 |
|---|---|---|
| 1. 자료와 모델이 보지 못한 것 | 경계 문단 · 발전량은 전력시장 거래분만이다(자가용 · PPA) · 모델은 LNG를 바로 움직이는 여섯을 보지 않는다 · 석탄 발전소 정비는 보지 않았다 · 1단계 짐작 ⑤의 물음 · 우리 LNG 값은 두 달쯤 늦게 따라온다 · 원전 설비용량은 멈춘 원전까지 센 값이다 · 발전소마다의 자료가 없다 | 7과 경계 문단 |
| 2. 자료가 적거나 좁은 것 | 기온은 서울 한 지점이다 · 학습에 쓴 해가 적다 · 2026년 7월은 비교에서 뺐다 · 아주 추운 시각은 드물다 · 시간별 약 3만 행 | 5 |
| 3. 모델과 채점 | 공휴일 · 원전 발전량을 입력에서 뺐다 · 평소도 해마다 달라진다 · 4단계에서 2023 · 2025년을 나란히 본 셀(미리 본 LNG) · 랜덤포레스트 설정 · 잔차가 이웃한 시각끼리 이어진다 · 넣지 않은 다섯 · 늘어난 태양광 설비 · 30도 넘는 시각 · 습한 날 · 원전 정비 일정과 석탄 · 크게 틀린 날이 이어졌다 · 주말에 든 공휴일 · 토요일 · 고른 모델의 2025년 성적 · 모델 넷을 평균 내지 않았다 · 모델 넷과 기준선이 같은 곳에서 틀렸다 · 선형회귀 0 ~ 9시 · 기준선보다 더 틀린 달 · 고르는 기준에 동의한 때 · 핵심인자의 순서 | 21 |
| 4. 결과를 읽을 때 | 배 몇 척 · 석탄이 많았던 까닭 · 전력 수요 · 그 밖의 발전원 · 배출권 비용 · 서로 다른 모델로 냈다 · 평소 범위로는 작은 변화를 잡지 못한다 · 날 종류 기준 · 9 ~ 11월에 몰린 날 · 3월 16일 앞뒤 비교 | 10 |
| 다른 칸으로 옮김 | 예측 대상은 회사 조달 물량이 아니다 · 가격은 예측하지 않는다(→ 「문제」의 하지 않은 것) · 받은 자료에서 뺀 기간이 셋이다(→ 「데이터」) | 3 |
| 합계 | | 46과 경계 문단 |

묶은 판은 서른아홉 줄입니다. 같은 것을 말하는 줄은 하나로 합쳤고(발전소마다의 자료 · 석탄 발전소 정비 · 원전 설비용량 → 한 줄, 공휴일 · 주말에 든 공휴일 → 한 줄, 선형회귀 0 ~ 9시 · 기준선보다 더 틀린 달 → 한 줄, 랜덤포레스트 설정 · 고른 모델의 2025년 성적 · 모델 넷을 평균 내지 않았다 · 고르는 기준에 동의한 때 → 한 줄), 경계 문단은 세 줄로 나눴습니다. 경계 문단의 첫 문장은 3-사의 그림 아래 줄과 같은 말이라 그 그림을 가리키게 고쳤습니다. 줄 안의 숫자와 근거는 앞 단계 📌 결론 칸과 지금 「한계」에 적힌 것만 썼고, 「가르다」는 「구별하다」로 바꿨습니다.

**README에 넣을 글** — 3-사의 저량 · 유량 그림과 그 아래 줄 다음에 둡니다.

> **한계**
>
> **1. 자료와 모델이 보지 못한 것**
>
> - **이 프로젝트가 자료로 가진 것은 LNG 발전량뿐이다(10단계 물음 3).** 위 그림의 들여온 LNG와 LNG 재고를 보지 못해, 2026년에 LNG가 모자랐는지는 말할 수 없다. 재고가 버틴 뒤 7월 이후에 나타났을 영향도 말할 수 없다. 2026년에 모델 예측과 달랐던 원전 · 석탄 · 전력 수요와 그 까닭 후보(원전은 2026년이 되기 전에 멈춘 원전 넷, 석탄은 계절관리제와 3월 16일 상한 해제 · 1 ~ 4월에 평소 해보다 작았던 연료비단가 차이 · 정부의 LNG 수요 관리)는 모두 모델에 넣지 않은 것이다.
> - **LNG 값은 LNG 발전량으로 바로 이어지지 않는다.** 한국 전력시장은 연료비가 싼 발전소의 전기부터 쓰고 LNG는 그러고 남는 만큼 발전해서, 석탄이 LNG보다 싼 동안은 LNG 값이 올라도 쓰는 순서가 바뀌지 않는다(2023 ~ 2026년 1 ~ 6월의 모든 달에 유연탄이 LNG보다 쌌다 — 10단계 2-가). 그래서 인과지도에서 이 화살표를 지웠고(7단계), LNG 값은 석탄 발전 상한과 전기 요금을 거쳐서만 LNG 발전량으로 이어진다. 보는 기간의 사정 때문이 아니라 시장의 구조 때문이다.
> - **모델은 LNG 발전량을 바로 바꾸는 것을 거의 보지 않는다(6 · 7단계 인과지도).** LNG 발전량으로 바로 들어오는 화살표는 전력 수요 · 원전 · 석탄 · 전력시장 태양광 · 그 밖의 발전원 · 추운 정도 여섯이고, 모델이 입력으로 보는 것은 추운 정도 하나다. 전력 수요 · 원전 · 석탄 · 전력시장 태양광은 모델이 보는 날씨 · 시각 · 날 종류를 거쳐 평소에 바뀌던 만큼만 알고, 그 밖의 발전원은 그것도 모른다(7단계 4-마). 그래서 2026년 차이에는 이것들이 평소 해와 다르게 움직인 만큼이 같이 들어 있다.
> - **발전소마다의 자료가 없다.** 원전마다의 정비 일정은 한수원 API를 불러 봤지만 날짜와 상관없이 최근 엿새만 줘서 쓰지 못했고(2a단계), 석탄 발전소마다의 정비 일정은 찾지 않았다 — 받은 자료는 석탄을 연료별 합계로만 주고, 석탄 설비용량은 한 해 내내 거의 같은 값이다(10단계 물음 2). 원전 설비용량도 멈춘 원전까지 센 값이라, 2023 → 2025년에 설비용량은 늘었어도 실제로 발전할 수 있던 원전은 오히려 줄었다(판단 — 9단계 4-다). 그래서 원전 · 석탄 발전량이 달마다 왜 바뀌었는지(정비인지)는 판단으로만 말하고, 석탄 발전소 정비는 까닭 후보로 넣지 않았다. 9단계 4-다에서 원전 발전량을 4월 4일 · 11일 앞뒤로 비교한 것도 원전 전체를 합한 값이라 다른 원전의 정비가 섞인다. 발전기마다의 운전 기록도 없어서, 2026년 주말에 석탄 발전기를 어떻게 운전했는지는 보지 못했다(주말 석탄 발전량이 모델 예측보다 적었던 것이 6 ~ 15시였다는 데까지 봤다 — 9단계 4-바). 원전은 발전량이 수요를 따라가지 않는다는 데까지 봤다(4단계).
> - **우리 LNG 값(EPSIS 연료비단가)은 두 달쯤 늦게 따라오는 값이다(10단계 2-라).** 늦게 따라온다는 것은 기사 하나(에너지경제신문 2020-10-22)로 본 것이다. 그날그날의 시장 가격은 받지 않아서, 3월 16일 석탄 상한 해제 때 시장의 LNG 값이 이미 올랐는지는 이 자료로 볼 수 없다(인과지도 ⑤ — 못 봄).
> - **발전량은 전력시장에서 거래된 것만이다.** 지붕 · 공장에서 바로 쓰는 태양광(자가용 · PPA)은 빠져 있다. 2025년에 전력시장 수요는 밤보다 낮에 모델 예측보다 더 적었지만(선형회귀로 나눈 날마다 낮 − 밤 −529 MWh — 7단계 4-다), 해가 강한 날일수록 더 적지는 않았다. 시장 밖 태양광이 늘어서인지, 낮에 전기를 쓰는 쪽(공장 · 상가)이 바뀌어서인지는 받은 자료로 구별할 수 없다(인과지도의 「시장 밖 태양광 → 전력 수요」 — 못 봄).
> - **카타르에서 오는 LNG가 줄었다면 어디서 가져왔을까(1단계 짐작 ⑤의 물음)는 답할 수 없다.** 자료에는 발전소가 쓴 LNG만 있고 들여온 LNG는 없다. 같은 줄의 짐작(석탄이나 원자력에서 가져올 것이다)은 석탄만 같은 쪽이었지만, 그 크기는 평소 해에도 있던 정도였고 사태 전인 1월 하순부터 많아서 사태 때문인지는 구별할 수 없다. 원전은 1 ~ 6월 내내 모델 예측보다 적었다(11단계 1-나).
> - **시스템 다이내믹스는 그림까지만 했다.** 인과지도로 모델이 보는 것과 보지 못하는 것을, 저량 · 유량 그림으로 자료가 가진 것과 가지지 못한 것을 나눠 그렸고, 시뮬레이션은 하지 않았다.
>
> **2. 자료가 적거나 좁은 것**
>
> - **배운 해가 적다.** 2025년을 채점할 때는 2023 · 2024년 두 해(계절이 두 번 반복된 자료)로, 2026년을 맞힐 때는 2023 ~ 2025년 세 해로 배웠다. 2024년 8월이 빠져 있어 8월은 다른 달보다 적은 해로 배웠다(「데이터」).
> - **기온 · 일사는 서울 한 지점의 값이다.** 전국의 날씨를 서울 한 곳으로 대신했다(4단계에서 지점을 늘리지 않기로 했다).
> - **아주 추운 시각은 드물다(4단계).** 서울에서 영하 12도 이하인 시각은 2023년 64시간 · 2024년 10시간 · 2025년 2시간뿐이라, 아주 추운 날의 예측은 이 적은 시각으로 배운 값이다. 영하 5도 이하 시각은 2025년에 잘 맞혔다(7단계 3-바).
> - **2026년은 1 ~ 6월만 비교했다(9단계 물음 1).** 받은 7월 값이 EPSIS 월 합계와 0.99% 어긋나서(평소 해에 가장 크게 어긋난 달은 0.24%) 넣지 않았다. 그래서 호르무즈 봉쇄(3월 4일) 뒤로는 넉 달만 봤다.
> - **자료는 시간별로 약 3만 행이다(2023 ~ 2026년 7월).** 그래서 이 프로젝트를 「빅데이터」라고 부르지 않는다. AI라고 부를 수 있는 것은 예측 모델뿐이다.
>
> **3. 모델과 채점**
>
> - **원전 발전량을 입력에서 뺐다(6단계 물음 1).** 배운 해에 석탄이 줄면서 원전과 LNG가 같이 늘어서, 넣으면 「원전 발전량이 줄면 LNG 발전량도 준다」로 반대로 배웠고 2024년도 더 못 맞혔다. 그래서 2026년 차이에는 원전 발전량이 모델 예측보다 적었던 영향(LNG를 늘리는 쪽)과 석탄 발전량이 많았던 영향(줄이는 쪽)이 같이 들어 있다. 그래디언트 부스팅은 둘을 나누지 못해, 선형회귀로 나눠 옆에 놓았다(9단계 2-나). 원전과 석탄이 따로 움직인 해가 배우는 자료에 들어오면 다시 볼 수 있다.
> - **넣어 보지 않은 것이 다섯 있다(7단계 3-가).** 부산 기상 · 일사를 달마다 따로 · 월 × 시간대 · 불쾌지수 · 태양광 설비를 아는 열이다. 앞 단계에서 「2025년에 이렇게 틀리면 다시 본다」고 미뤄 두었지만, 2025년을 보고 넣으면 2025년 점수가 처음 보는 해의 점수가 아니게 되어 넣지 않았다.
> - **잔차가 이웃한 시각끼리 이어진다(7단계 물음 1).** 모델이 틀린 양(잔차)은 이웃한 시각끼리 같은 쪽으로 이어진다(선형회귀에서 한 시각 떨어진 잔차끼리 상관 0.92). 그래서 서로 독립인 자료 수는 시각 수보다 훨씬 적고, 보통 방법으로 낸 p값은 실제보다 작게 나온다. 평소 범위는 공식으로 내지 않고 2024 · 2025년에 실제로 틀린 크기로 잡았다(공식으로 내면 열 배 넘게 좁게 잡힌다 — 7단계 3-더).
> - **2025년에 틀린 곳은 모델 넷과 기준선이 같았다(7단계 물음 3).** 날씨를 아는 모델 넷과 날씨를 모르는 기준선이 12달 가운데 11달에서 같은 쪽으로 틀렸다. 2025년이 2023 · 2024년과 달랐던 것의 많은 부분은 모델에 넣은 것(날씨 · 시각 · 날 종류)으로는 설명되지 않는다고 봤고(판단), 확인한 까닭은 아래 줄들과 인과지도의 화살표로 적었다(7단계 물음 4).
> - **늘어난 태양광 설비를 따라가지 못했다(7단계 3-라 · 3-마 · 3-파).** 2025년에 모델 넷 모두 밤보다 낮에 LNG 발전량을 실제보다 많게 예측했다(날마다 낮 − 밤 −822 ~ −1,164 MWh). 같은 햇빛에서 2025년 태양광 발전량이 2023 · 2024년보다 한 시각 698 ~ 1,049 MWh 많아서(설비용량 평균 7,840 → 8,998 → 10,072 MW), 늘어난 설비로 설명된다고 본다(판단). 모델은 햇빛 세기만 알고 설비가 는 것은 모른다. 선형회귀 둘은 봄(3 · 5월) 낮에 이 치우침이 트리 계열보다 훨씬 컸다(햇빛의 효과를 한 해 내내 같은 크기로 배워서로 본다 — 판단).
> - **30도 넘는 시각에 가장 크게 틀렸다(7단계 3-바).** 2025년에 모델 넷 모두 이 시각에 LNG 발전량을 실제보다 한 시각 371 ~ 997 MWh 많게 예측했고, 선형회귀 둘은 기준선보다도 더 틀렸다. 30도 넘는 시각은 두 해 597시각(3.6%)뿐이라, 30도에서 한 번 더 나누는 열을 넣어도 결정계수가 0.001만 올랐다(5단계). 서울 밖(부산) 기온이 없어서 틀린 것인지는 부산 자료를 받지 않아 구별할 수 없다.
> - **습한 날에는 덜 습한 날보다 LNG 발전량을 실제보다 적게 예측했다(7단계 3-사 · 3-차).** 2025년 6 ~ 9월에 같은 달 · 같은 시각끼리 비교하면, 습한 날은 덜 습한 날보다 「실제 − 모델 예측」이 한 시각 266 ~ 551 MWh 컸다. 차이가 해가 강한 한낮에 가장 커서 흐린 날 태양광이 적었던 쪽으로 읽었고(판단), 습도 때문에 냉방이 늘었다는 것은 확인되지 않았다.
> - **원전과 석탄 발전량이 바뀐 것을 모델은 모른다(7단계 3-아 · 3-하 · 3-거).** 2025년에 달마다 모델이 틀린 쪽은 원전만으로는 12달 가운데 5달, 석탄을 더하면 10달이 설명됐다. 1 ~ 4월은 원전 발전량이 2023 · 2024년보다 많았는데도 석탄 발전량이 24 ~ 33% 적어서 LNG 발전량이 모델 예측보다 많았다. 남은 두 달(5 · 8월)은 수요가 모델 예측과 달랐던 만큼으로 설명됐다.
> - **크게 틀린 날은 며칠씩 이어져 나왔다(7단계 3-자 · 3-거).** 2025년에 가장 크게 틀린 열 날 가운데 7월 18 · 19일, 9월 11 · 13 · 14일이 붙어서 나왔다. 선형회귀로 나눠 보면 그날 틀린 양도 대부분 원전 · 석탄 발전량이 모델 예측과 달랐던 만큼이었고, 그다음이 수요였다. 왜 그렇게 움직였는지(발전소 정비 · 정책 · 서울 밖 날씨)는 받은 자료에 없다.
> - **주말에 든 공휴일과 연휴에 LNG 발전량을 실제보다 많게 예측했다(5단계 물음 5 · 7단계 3-자).** 주말에 든 명절 아닌 공휴일은 실제가 모델 예측보다 한 시간에 약 2,000 MWh 적었다(2023 · 2024년 사흘의 평균이고, 날마다 방향이 다르다 — 5단계 선형회귀). 2025년에도 3월 1일(토) · 2일(일)이 크게 틀렸다(대체 휴일까지 사흘 연휴). 5월 1일은 2023 · 2024년에 실제가 모델 예측보다 적었지만 2025년에는 크게 틀리지 않았다. 명절은 두 해의 설 · 추석 네 번으로 배운 값이다.
> - **토요일에 더 틀렸다(7단계 3-카).** 2025년에 가장 크게 틀린 37일 가운데 9일이 토요일이었고(요일과 관계없으면 5.3일), 다섯 방법 모두 토요일의 평균 절대 오차가 가장 컸다. 37일 가운데 9일은 우연으로도 7%쯤은 나오는 정도라 약하게만 말한다.
> - **선형회귀는 일부 시간대와 달에 기준선보다 더 틀렸다(7단계 3-다 · 8단계 물음 1).** 2025년 0 ~ 9시에 기준선보다 더 틀렸고(0 ~ 5시 한 시각 2,887 대 기준선 2,492 MWh — 기온 × 시간대 선형회귀도 0 ~ 5시에는 2,594), 달마다 보면 일곱 달(1 · 3 · 6 · 7 · 8 · 11 · 12월)에 기준선보다 더 틀렸다. 랜덤포레스트는 12월에 더 틀렸고, 고른 그래디언트 부스팅은 기준선보다 더 틀린 달이 없었다. 2025년을 본 뒤라 고치지 않았다.
> - **평소도 해마다 조금씩 달라진다(6단계 1-다).** 2023년으로 배운 모델 셋 모두 2024년에 실제가 모델 예측보다 한 시각 평균 821 ~ 895 MWh 많았다. 그래서 2026년 차이는 2024 · 2025년에 모델이 틀린 크기(평소 범위)와 나란히 놓고 읽었다.
> - **모델을 고른 점수는 조금 좋게 나온 값일 수 있다(6 · 8단계).** 랜덤포레스트 설정은 2024년 결정계수로 골라서, 2024년 점수는 조금 좋게 나왔다(바꾸기 전 설정보다 2024년에 +0.0083 높던 것이 2025년에는 +0.0006이었다 — 7단계 2-다). 모델 넷 가운데 2025년에 가장 나은 것을 골랐으니, 고른 모델의 2025년 점수에도 운이 좋았던 만큼이 조금 섞인다(판단). 고르는 방법 둘은 8단계 결과를 보기 전에 정했지만 7단계에서 2025년 결과를 본 뒤였고, 그 가운데 「같은 달이 두 해 모두 기준선보다 더 틀린 모델을 뺀다」는 덜 엄격한 쪽으로 정한 것에 클로드가 동의한 것은 결과를 본 뒤였다(더 엄격하게 셌어도 같은 모델이 남았다). 모델 넷을 평균 내서 맞히는 방법은 6단계에서 정하지 않아 해 보지 않았다.
> - **핵심인자의 순서는 미리 기준으로 정한 셀이 아니라 다른 셀에서 가져왔다(8단계 물음 2).** 미리 정한 셀(2-나 — 그 변수만 흔히 있는 값으로 바꿨을 때 예측이 바뀐 폭)은 햇빛만 낮 10 ~ 15시로 폭을 구해서 다섯 변수를 같은 시각 범위로 비교하지 못했다. 그래서 순서는 2-다(그 변수의 값만 다른 시각의 값으로 바꿔 놓았을 때 모델이 더 틀린 양)에서 가져왔는데, 2-다도 새벽 2시에 한낮의 햇빛 값이 붙는 것처럼 실제로는 없는 조합을 만든다. 어느 셀로 적을지는 두 셀의 순서를 본 뒤에 정했다.
> - **2025 · 2026년 LNG를 미리 본 적이 있다.** 2단계에서 자료가 맞는지 EPSIS와 맞춰 본 월 합계, 4단계에서 2023년과 2025년을 나란히 본 셀, 09-15에 본 2026년 1 ~ 6월 합계(2025년보다 −1.0%), 09-17에 옛 6단계 재료로 낸 2025년 기준선 오차와 LNG 비중이다. 넣을 열과 모델 후보는 2023년으로 배워 2024년을 맞힌 숫자로만 정했고, 2025년 LNG는 7단계 채점과 그 뒤(모델 고르기 · 평소 범위)에, 2026년 LNG는 9 · 10단계 비교에만 썼다.
>
> **4. 결과를 읽을 때**
>
> - **평소 범위로는 작은 변화를 잡지 못한다(9단계 물음 2 · 3).** 달 단위 기준(한 시각 ±2,177 MWh)은 한 달 LNG 발전량의 10 ~ 15%보다 큰 변화만 잡고, 날 단위 기준으로는 몇 주에 걸친 작은 변화를 잡지 못한다(주 단위 기준은 두지 않았다). 그래서 「달랐다고 말할 근거가 없다」는 「같았다」는 뜻이 아니다.
> - **「얼마나 달랐나」와 「무엇이 달랐나」는 서로 다른 모델로 냈다(9단계 2-나).** LNG 발전량이 평소 범위 안인지는 그래디언트 부스팅으로, 원전 · 석탄 · 전력 수요의 「실제 − 모델 예측」과 석탄의 비율은 선형회귀로 냈다. 선형회귀는 차이를 수요 · 원전 · 석탄 · 그 밖의 넷으로 나누면 그 합이 차이와 꼭 맞아서 나눠 읽을 수 있고, 그래디언트 부스팅은 그렇게 나눌 수 없다.
> - **날 종류로 나눈 기준은 약하다(9단계 3-마).** 2024 · 2025년 가운데 더 크게 틀린 해의 크기를 기준으로 삼아서, 날 종류 일곱 줄 가운데 2.3줄쯤은 우연으로도 넘는다(어림 — 판단). 2026년 토 · 일이 그 기준을 넘은 것은 이 약점과 같이 읽는다.
> - **2024 · 2025년에 크게 틀린 날은 9 ~ 11월에 몰려 있었다(9단계 3-가).** 그래디언트 부스팅이 가장 크게 틀린 열 날 가운데 여섯이었다(날 수대로라면 2.6날). 2026년 비교는 1 ~ 6월이라 그 까닭은 보지 않았다.
> - **석탄 발전량이 2026년에 모델 예측보다 많았던 까닭은 찾지 못했다(10단계 물음 2).** LNG와 석탄의 연료비단가 차이로는 설명되지 않는다(연료비단가에는 배출권 비용과 연료에 붙는 세금이 들어 있는 것으로 본다 — 판단. 세금이 들어 있고 국내탄 발전기는 지원금을 뺀 값을 쓴다는 것은 비용평가 세부운영규정으로 확인). 석탄과 LNG의 발전량이 각각 모델 예측보다 많았던 양을 합친 것 가운데 석탄의 비율은 1월 1%에서 5월 91%로 커졌다(선형회귀). 시기가 맞는 후보 셋(계절관리제와 3월 16일 상한 해제 · 1 ~ 4월에 평소 해보다 작았던 연료비단가 차이 · 정부의 LNG 수요 관리) 가운데 무엇이 까닭인지는 구별할 수 없고, 상한이 석탄을 줄이고 있었는지도 구별하지 못했다(9단계 4-나). 석탄의 비율은 2월에 이미 36%였는데, 2월에 걸리는 후보(계절관리제 · 연료비단가 차이)는 비율이 1%였던 1월에도 같았고 계절관리제는 석탄을 줄이는 쪽이라, 2월에 커진 까닭은 설명되지 않는다(판단). 6월에 석탄 발전량이 많아서 LNG 발전량이 적었다고 잇는 것도 판단이다. 10단계 2-다 ~ 2-아는 2-가 · 2-나의 결과를 본 뒤에 연 셀이다.
> - **전력 수요(전력시장에서 거래된 양)도 2026년 1 ~ 4월과 6월에 모델 예측보다 적었다(10단계 2-아).** 전기를 덜 썼는지, 전력시장 밖의 태양광 발전이 늘었는지는 이 자료로 구별할 수 없다.
> - **5 · 6월에 그 밖의 발전원이 모델 예측보다 덜 발전한 양의 대부분은 신재생 · 기타였다(10단계 2-바 · 2-사).** 그 가운데 태양광보다 나머지(풍력 · 바이오 · 연료전지 같은 것)가 컸지만, 나머지 가운데 무엇인지는 2026년 자료로 나눌 수 없다. 2026년 태양광은 신재생 · 기타에서 그날 밤 값을 뺀 어림이라 실제보다 조금 크게 나오고(2024년에 맞춰 보면 3 ~ 10월 −0.7 ~ +7.8%, 11 ~ 2월 +8.3 ~ +13.0%), 태양광이 적었던 것이 서울 밖 날씨 때문인지 출력 제어 때문인지는 구별할 수 없다. 2023년 연간 파일에는 소수력이 따로 없어서 수력에 들어 있을 수 있다(그렇다면 10단계 2-바의 수력 칸이 조금 낮게, 신재생 · 기타 칸이 그만큼 높게 나온다 — 어림).
> - **배출권 비용을 넣은 값 차이는 따로 계산하지 않았다(10단계 물음 2).** 전력거래소는 배출권 비용을 발전사가 그 달 실제로 사고판 돈으로 계산해 다음 달 열량단가에 넣는데(비용평가 세부운영규정 2025. 12.), 그 기록은 공개 자료에 없다. 넣으면 석탄이 더 비싸지는 쪽이라 「값 차이로 설명되지 않는다」는 그대로이고, 발전 부문은 2026년에도 배출권의 85%를 무상으로 받아(계획 — 9단계 4-가) 그 크기도 작다고 본다(판단).
> - **석탄 발전 상한 해제(3월 16일) 앞뒤 비교는 날씨를 맞추지 않았다(9단계 4-나).** 실제 발전량끼리 비교해서 해마다 다른 날씨가 섞여 있다. 날씨를 맞춘 9단계 4-가(주마다 석탄 발전량의 「실제 − 모델 예측」)에서도 2026년 석탄은 3월 16일 뒤에 오르지 않았다.
> - **배 몇 척은 범위로만 말할 수 있다(10단계 물음 1).** 배 한 척은 약 356 ~ 497 GWh로 잡았다. 효율(한국전력통계 2023 ~ 2025년 평균 45.46%)이 어느 발열량 기준인지 확인되지 않은 만큼과 배 크기(57,000 ~ 72,000톤)에서 생긴 범위다. 이 효율은 한전 자회사 복합화력 평균으로 보이는 값이라 민간 발전사까지 든 우리 자료와 대상이 조금 다르고, 모델 예측과 달랐던 만큼의 전기는 평균보다 효율이 낮은 발전소에서 나와 실제 척 수는 범위의 많은 쪽보다도 많을 수 있다(판단). 톤과 척은 발전에 쓴 LNG를 바꾼 양이지, 들여온 LNG나 배의 수가 아니다. 효율의 한쪽 끝은 처음에 프랑스 발전소의 시험 기록으로 두었다가, 한국 자료에 맞지 않아 뺐다.

답: 

## 3-자 — README 맨 위 「한눈에 보기」

**이 셀이 답하는 물음** — 처음 보는 사람이 3분 안에 이 프로젝트의 물음 · 한 일 · 결과 · 한계를 알 수 있게 쓰면 어떻게 되나(창주 님 한 줄의 「프로젝트를 처음 본 사람이 이해할 수 있도록」 · 물음 3 머리 칸의 3분 설명).

**어디서 나온 셀인가** — 3-가에서 정한 차례의 첫 칸입니다. 3-아에서 핵심본을 둘로 나누기로 해서, 이것은 다른 사람에게 보여 주는 3분짜리이고 창주 님이 보시는 10분짜리는 따로 만듭니다. 결과 글(물음 1)과 묶은 「한계」(3-아)에서 뽑았습니다. 채팅에서 먼저 보여 드렸고, 「결과가 조금만 더 상세했으면 좋겠고. 나머진 괜찮은거 같어.」라고 하셔서 결과를 늘렸다가 「이건 좀 너무 많은데..?? 이건 다른 사람에게 보여주는 버전인가?? 내가 보는거??」라고 하셔서, 처음 판의 결과에 세 가지(날 · 낮밤으로 나눠서도 봤다는 것, 평소 범위가 배 몇 척인지, 석탄의 비율이 커진 것)만 더했습니다.

**README에 넣을 글**

> **한눈에 보기**
>
> **물음** — 2026년 1 ~ 6월(호르무즈 사태가 있던 때) 우리나라 LNG 발전량은, 날씨 · 시각 · 날 종류가 같을 때의 평소 패턴과 얼마나 달랐나.
>
> **한 일** — 전력거래소의 시간별 발전량과 기상청 서울 날씨(2023 ~ 2026년, 시간별 약 3만 행)로, 몇 시인가 · 무슨 날인가 · 기온 · 햇빛 · 몇 월인가를 보고 LNG 발전량을 예측하는 모델을 만들었다. 2023 · 2024년으로 배운 모델 넷을 2025년으로 채점해 그래디언트 부스팅을 골랐고(결정계수 0.7553, 날씨를 모르는 기준선보다 나았다), 2023 ~ 2025년으로 다시 배워 2026년 1 ~ 6월을 예측해 실제와 비교했다.
>
> **결과**
> - 2026년 1 ~ 6월 LNG 발전량은 달마다 모두 평소 범위 안이었다. 평소 범위는 모델이 2024 · 2025년을 예측했을 때 한 달 평균으로 가장 크게 틀린 크기(한 시각 ±2,177 MWh — 발전에 쓴 LNG를 배로 바꾸면 30일에 약 3 ~ 4척)다. 날로 나눠 봐도, 낮과 밤으로 나눠 봐도 2024 · 2025년에 가장 크게 틀린 날 · 달보다 크게 틀린 곳은 없었다. 그래서 평소 패턴과 달랐다고 말할 근거가 없다(같았다는 뜻은 아니다).
> - 그 안에서 LNG 발전량은 1 ~ 3월에 모델 예측보다 많았고 6월에는 적었다. 원전 발전량은 1 ~ 6월 내내 모델 예측보다 적었고(LNG를 늘리는 쪽), 석탄 발전량은 2월부터 많았고 전력 수요는 5월을 뺀 달에 적었다(줄이는 쪽). 이 둘이 같은 달에 상쇄되어 LNG의 차이가 작게 나왔다. 석탄과 LNG의 발전량이 각각 모델 예측보다 많았던 양을 합친 것 가운데 석탄의 비율은 1월 1%에서 5월 91%로 커졌다.
> - 석탄 발전량이 많았던 까닭은 찾지 못했다. LNG와 석탄의 연료비단가 차이로는 설명되지 않고, 시기가 맞는 후보(계절관리제와 3월 16일 상한 해제 · 연료비단가 차이 · 정부의 LNG 수요 관리)는 셋이지만 이 자료로는 구별할 수 없다. 석탄이 많아지기 시작한 때(1월 하순)는 호르무즈 사태(2월 28일)보다 앞이었다.
> - LNG 발전량을 크게 움직인 순서는 몇 시인가 → 무슨 날인가 → 기온 → 햇빛 → 몇 월인가였다.
>
> **한계** — 자료로 가진 것은 LNG 발전량뿐이라, 들여온 LNG와 재고를 보지 못해 2026년에 LNG가 모자랐는지는 말할 수 없다. 평소 범위는 한 달 LNG 발전량의 10 ~ 15%보다 작은 변화는 잡지 못한다.
