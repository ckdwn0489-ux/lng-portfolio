# 2단계 수집 — 공휴일 (한 번만 실행하면 끝)

**무엇을 하나** — 2023년부터 2026년까지의 한국 공휴일 날짜를 만들어 `data/raw/holidays_kr.csv`로 저장한다.

**왜 받는가** — 1단계에서 적은 입력 후보 여덟 가운데 하나가 공휴일이다. 공장이 쉬면 전기를 덜 쓰고, 그러면 LNG 발전소도 덜 돈다.

**어디서 받는가** — 인터넷에서 내려받는 것이 아니라 `holidays`라는 파이썬 패키지가 날짜를 계산해 준다. 설날·추석처럼 음력으로 정해지는 날과 대체공휴일까지 넣어 준다.

**5단계에서 정할 것** — 이 목록을 입력에 넣을지 말지, 그리고 이 중 무엇을 공휴일로 볼지는 5단계에서 정한다. 여기서는 받아 두기만 한다.
---

**어떤 흐름에서 이 노트북이 만들어졌나 (2026-09-19)**

2단계를 닫기 전에 「2단계에서 안 한 것이 남았나」를 점검했다. 그때 클로드가 「공휴일은 내려받을 파일이 아니라 패키지 설치라, 실제로 쓰는 5단계에서 깔면 된다」고 적었다. 창주가 「공휴일 자료도 받으면 되는 것 아닌가, 5단계에서 다시 돌아와서 하는 흐름인가」라고 물었고, 지침에 「수집은 2단계 몫」이라 적혀 있어 클로드 판정을 뒤집고 2단계에서 받기로 했다. 5단계에서 정하는 것은 「넣을까 말까」이지 「받을까 말까」가 아니다.


```python
# 전체 11단계 중 → 2단계 데이터 수집과 구성 보기 · 2023~2026년 한국 공휴일을 만들어 파일로 저장한다
import subprocess, sys                              # 패키지가 없으면 깔려고 쓴다
try:
    import holidays                                 # 공휴일 날짜를 계산해 주는 패키지
except ImportError:                                 # 안 깔려 있으면
    subprocess.run([sys.executable, "-m", "pip", "install", "-q", "holidays"])   # 조용히 깐다
    import holidays                                 # 깔고 나서 다시 불러온다

import os                                           # 폴더 경로를 다루는 기본 모듈
import pandas as pd                                 # 표를 다루는 라이브러리

ROOT = os.getcwd()                                  # 지금 폴더(보통 notebooks/)에서 시작해
while not os.path.isdir(os.path.join(ROOT, "data")) and ROOT != os.path.dirname(ROOT):
    ROOT = os.path.dirname(ROOT)                    # data 폴더가 보일 때까지 위로 올라간다
저장할곳 = os.path.join(ROOT, "data", "raw", "holidays_kr.csv")

kr = holidays.KR(years=range(2023, 2027))           # 2023~2026년 한국 공휴일
표 = pd.DataFrame(sorted(kr.items()), columns=["date", "name"])    # 날짜순으로 표를 만든다
표["date"] = pd.to_datetime(표["date"])              # 글자를 날짜로 바꾼다
표.to_csv(저장할곳, index=False)                      # 파일로 저장한다

print("저장했다:", 저장할곳)
print("전체", len(표), "일\n")
print("해마다 며칠인가")
print(표.groupby(표["date"].dt.year).size().to_string())
```

```python
# 전체 11단계 중 → 2단계 데이터 수집과 구성 보기 · 2026년 공휴일 목록을 눈으로 본다 · 앞 셀을 먼저 실행한 뒤에 돌린다
올해 = 표[표["date"].dt.year == 2026].copy()          # 2026년만 골라낸다
올해["요일"] = 올해["date"].dt.day_name()              # 무슨 요일인지 붙인다
올해["date"] = 올해["date"].dt.strftime("%m-%d")                  # 월-일만 보이게
print("2026년 공휴일", len(올해), "일 — 적용 구간은 1~6월이다\n")
print(올해.to_string(index=False))
```
