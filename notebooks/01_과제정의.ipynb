## 📌 물음 1 결론

**왜 생긴 물음인가** — 09-02에 세운 첫 설계(장기계약과 스팟 중 어느 쪽이 유리했나)는 맞춰 볼 정답이 없고 두 가격의 뺄셈이라 09-12에 접었다. 새 설계로 무엇을 묻고 무엇을 맞힐지 다시 적은 물음이다.

### 결정 — 물음은 「2026년 LNG 발전량은 평소 패턴에서 얼마나 벗어났는가?」, 맞힐 값은 시간별 LNG 발전량(MWh)

**같이 정한 것** — ① 그 차이에서 주요 원인을 살펴보는 것까지 물음에 넣었다. ② Why는 「예측하는 모델을 만들고, 실제와 비교한 뒤, 실제 원인을 데이터에서 찾아낸다」로 적었다. ③ 과제명은 그대로 둔다.

**결정 근거** — 09-12에 세운 기준 셋을 따랐다. ① 맞춰 볼 정답이 데이터에 있어야 한다(시간별 LNG 발전량은 전력거래소 자료에 있다). ② 뺄셈으로 나오는 건 프로젝트가 아니다. ③ 범위를 스스로 넓히지 않는다. 판단은 「✍️ 물음 1 — 내가 정한 것」 ①~④에 적었다.

**남은 것** — 원인을 어디까지 말할 수 있는지는 3단계 물음 2의 2-다 「✍️ 읽고 나서」 아래(09-19)에 클로드가 좁혀 적었다. 모델이 모르는 것(석탄·가격·수요 변화)이 차이에 함께 섞이므로, 쓸 수 있는 문장은 「모델이 아는 조건(기온·시간대·요일·원전·태양광)으로 설명되지 않는 차이가 이만큼이다」이다. 그 옆에 석탄·가격을 놓고 읽는 것은 10단계(그 답을 숫자로 내는 단계)에서 한다.

→ **6단계를 닫으며 이어진 곳(10-05)** — 원전은 6단계 물음 1에서 입력에서 뺐고(10-03), 태양광은 5단계 물음 3에서 일사 한 열로 바꿨다(09-27). 그래서 위 문장은 「날씨·시간대·날 종류가 같을 때 평소 패턴과 비교해 2026년 LNG가 얼마나 달랐나」로 바뀌었다. 원전·석탄·가격은 9·10단계에서 그 차이 옆에 놓고 읽는다.

→ **10-13에 다시 짚으며 이어진 곳** — 8단계(모델을 고르고 핵심인자를 정하는 단계)를 하다가 흐름을 놓쳐, 1단계부터 기억으로 다시 답하셨다. 답하신 것에서 빠진 둘을 채웠다. ① 물음의 「어떤 변수가 원인인가」는 둘로 나뉜다 — 모델에 넣어 8단계에서 영향의 크기를 내는 것(몇 시인가 · 무슨 날인가 · 기온 · 햇빛 · 몇 월인가 — 2026년의 사정과 상관없이 주어지는 것)과, 모델에 넣지 않고 9 · 10단계에서 2026년 차이 옆에 놓고 읽는 것(원전 · 석탄 · 가격)이다. 원전 · 석탄 · 가격도 평소에 LNG를 움직이지만, 석탄 · 가격은 2026년에 달라진 것 자체라 넣으면 보려는 차이가 사라진다. 원전은 6단계 물음 1에서 넣어 보니, 2024년에 석탄이 줄면서 원전과 LNG가 같이 늘어 「원전이 늘면 LNG도 는다」로 거꾸로 배워서 뺐다. 석탄을 넣지 않는 까닭으로 기억하신 「넣으면 뺄셈이 된다」는 총 전력거래량과 다른 발전원을 함께 넣어 「LNG = 총 전력거래량 − 다른 발전원」이 될 때만 맞는 옛 설명이다(09-22에 고침). ② 정답(모델이 맞히도록 배우는 값)은 모델의 예측량이 아니라, 전력거래소에 실제로 기록된 시간별 LNG 발전량이다. 문답은 `docs/7단계_10-13_다시짚기_문답.md` 2절에 있다.

### 클로드 답 — 물음 1

맞은 것

- 「수치적인 데이터로 살펴보는 게 목적이야. 느낌이 아니라」 — 이것이 Why입니다. 제 초안과 같은 말이고 더 짧습니다.
- 과제명을 그대로 두기로 한 것. 지금 제목이 교재가 말하는 대로 되어 있습니다. 대상은 발전용 LNG 소요, 방법은 평소 패턴을 아는 모델, 효과는 2026년이 평소와 얼마나 다른지 확인한다입니다.

다르게 읽은 것

- 「각 요소가 어느 정도 비율로 영향을 미쳤나」는 절반만 답이 나옵니다. 입력에 든 것(기온·원전·태양광 설비용량)은 「그 값이 2025년 수준이었다면」으로 바꿔 넣고 다시 예측하면 그 요소가 설명하는 몫이 얼마인지 확인할 수 있습니다. 입력에 없는 것(가격, 석탄 발전 상한 해제)은 예측과 실제의 차이로 남을 뿐이라 원인별로 가르지 못합니다.
- 가르려면 석탄과 총 전력수요를 입력에 넣어야 하는데, 그러면 LNG = 총량 − 나머지가 되어 기준 ②에 걸립니다. 09-12에 첫 설계를 접은 이유와 같습니다.
- 「26년의 예상 발전량과 그 오차」에서 오차는 두 가지가 섞인 값입니다. 모델이 원래 틀리는 양과, 2026년이 평소와 다른 크기입니다. 그 둘을 가르는 숫자는 7단계에서 냅니다.

## 물음 2 — 무엇이 LNG 발전량에 영향을 줄 것 같은가

무엇이 LNG 발전량을 늘리거나 줄일 것 같습니까? 한 줄에 하나씩, 방향까지 적습니다.

예) 기온이 낮아지면 난방 수요가 늘어 LNG가 늘어난다
예) 원전이 줄어들면 그만큼을 LNG가 대신해서 늘어난다

- 데이터가 있는지 없는지는 여기서 적지 않습니다. 2단계에서 정합니다.
- 4단계에서 보고 알게 된 것은 줄 끝에 (이미 본 것)이라고 적습니다. 8단계에서 모델이 뽑은 변수와 대조할 때 따로 셉니다.
- 09-12에 제가 입력 후보로 고른 여덟은 기온 · 습도 · 일사 · 시간대 · 요일(주말) · 공휴일 · 같은 시각의 원전 발전량 · 태양광 설비용량입니다. 창주 님 목록을 먼저 적으신 뒤에 이 여덟과 대조하겠습니다. 제 후보에 빠진 것이 보이면 그것이 이 단계에서 나온 것입니다.

## 📌 물음 2 결론

**왜 생긴 물음인가** — 모델을 만들기 전에 LNG 발전량을 움직일 원인을 먼저 적어 두고, 8단계(모델을 확정하고 핵심 변수를 고르는 단계)에서 모델이 고른 변수와 대조하려고 연 물음이다. 2단계(09-12)와 옛 4단계(09-13)에서 데이터를 일부 본 뒤인 09-18에 적었다.

**결정 근거** — 데이터가 있든 없든 떠오르는 대로 적는 물음이라 셀이 없다. 여섯 줄은 「✍️ 물음 2 — 내가 정한 것」 1~6에 적었다 — 1. 석탄보다 탄소가 적어 LNG 비중이 커질 것이다 2. 추우면 난방 수요가 늘어 LNG가 는다 3. 발전용이면 계절 영향이 난방만큼 크지 않을 수 있다 4. 연료별 경제성이 무엇을 태울지 정한다 5. 카타르 물량이 지정학 위험을 받으면 석탄·원전이 대신한다 6. 결국 상대 가격이 정한다.

- 적으신 셋은 각각 답이 나오는 단계가 있습니다. 「모델이 얼마나 정확한가」는 6단계에서 지표와 기준선을 정하고 7단계에서 채점합니다. 「예측과 실제의 차이로 어떤 원인이 얼마나 영향을 끼쳤나」는 9·10단계입니다. 「어떤 기준으로 범위를 정하고 논리가 정당하다고 말할 수 있나」는 7단계의 잔차 가정 셋(등분산성·정규성·독립성)과, 6단계의 「지표는 결과를 보기 전에 정한다」입니다.
- 이 셋은 1단계를 닫을 때 6·7·9단계의 물음으로 handoff에 적어 두겠습니다. 그 단계를 열 때 물음 칸에 그대로 나옵니다.
